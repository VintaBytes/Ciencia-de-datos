# Lectura de archivos con Pandas

En este cuaderno vamos ver como cargar archivos con diferentes formatos y convertirlos en un dataframe de Pandas. Los archivos debes descargarlos de [este enlace](https://github.com/VintaBytes/Ciencia-de-datos/tree/main/cursos/machine-learning-python2/Clase01/datos)) y almacenarlos en el espacio interno de Google Colab:


- `ejemplo.txt`
- `ejemplo.csv`
- `ejemplo.json`
- `ejemplo.xls`

Se supone que los archivos fueron cargados previamente en la carpeta `/content/`, que es el directorio de trabajo habitual de Colab.

En cada caso, el archivo se convertirá en un **DataFrame de Pandas** y se mostrarán sus primeras cinco filas.


## 1. Archivo de texto de ancho fijo

Primero leeremos `ejemplo.txt`. Como se trata de un archivo de texto cuyas columnas están ubicadas en posiciones de ancho fijo, utilizaremos `pd.read_fwf()`.

El resultado se almacenará en un DataFrame llamado `df_txt`, y luego mostraremos sus primeras cinco filas.


```python
import pandas as pd

# --------------------------------------------
# Carga el archivo de texto de ancho fijo
# almacenado en el espacio interno de Colab.
# --------------------------------------------

ruta_txt = "/content/ejemplo.txt"

df_txt = pd.read_fwf(ruta_txt)

# Mostramos las primeras cinco filas del DataFrame.
df_txt.head()
```

### Sintaxis utilizada

`pd.read_fwf(ruta_txt)` lee un archivo de texto con **campos de ancho fijo**. La sigla `fwf` proviene de *fixed-width formatted*.

La variable `ruta_txt` contiene la ubicación del archivo dentro de Colab. El resultado de la lectura se guarda en `df_txt`, que es un DataFrame de Pandas.

Finalmente, `df_txt.head()` muestra, de manera predeterminada, las primeras cinco filas del conjunto de datos.


## 2. Archivo CSV

Ahora leeremos `ejemplo.csv`. Los archivos CSV almacenan los datos en forma tabular, normalmente separando los valores mediante comas.

Pandas dispone de la función `pd.read_csv()` para cargar directamente este tipo de archivo en un DataFrame.


```python
# --------------------------------------------
# Carga el archivo CSV almacenado
# en el espacio interno de Colab.
# --------------------------------------------

ruta_csv = "/content/ejemplo.csv"

df_csv = pd.read_csv(ruta_csv)

# Mostramos las primeras cinco filas del DataFrame.
df_csv.head()
```

### Sintaxis utilizada

`pd.read_csv(ruta_csv)` abre el archivo CSV indicado por `ruta_csv` y convierte sus filas y columnas en un DataFrame.

En este ejemplo no indicamos manualmente el separador porque Pandas supone que los valores están separados por comas.

El método `df_csv.head()` permite observar rápidamente las primeras cinco filas y comprobar que la importación se realizó correctamente.


## 3. Archivo JSON

A continuación cargaremos `ejemplo.json`. JSON es un formato muy utilizado para almacenar e intercambiar información estructurada.

Pandas puede interpretar determinados archivos JSON directamente mediante `pd.read_json()` y convertirlos en un DataFrame.


```python
# --------------------------------------------
# Carga el archivo JSON almacenado
# en el espacio interno de Colab.
# --------------------------------------------

ruta_json = "/content/ejemplo.json"

df_json = pd.read_json(ruta_json)

# Mostramos las primeras cinco filas del DataFrame.
df_json.head()
```

### Sintaxis utilizada

`pd.read_json(ruta_json)` lee el contenido del archivo JSON y trata de organizar sus datos en filas y columnas.

El resultado se guarda en `df_json`. Esto permite trabajar con la información del archivo utilizando las mismas herramientas de Pandas que utilizaríamos con cualquier otro DataFrame.

Con `df_json.head()` mostramos las primeras cinco filas del dataset obtenido.


## 4. Archivo de Excel XLS

Por último leeremos `ejemplo.xls`, un archivo correspondiente al formato clásico de Microsoft Excel.

Para trabajar con archivos `.xls`, Pandas puede necesitar la biblioteca `xlrd`. Por eso primero la instalaremos y luego utilizaremos `pd.read_excel()`.


```python
# --------------------------------------------
# Instala el motor necesario para leer
# archivos de Excel en formato .xls.
# Luego carga el archivo en un DataFrame.
# --------------------------------------------

!pip install -q xlrd

ruta_xls = "/content/ejemplo.xls"

df_xls = pd.read_excel(ruta_xls)

# Mostramos las primeras cinco filas del DataFrame.
df_xls.head()
```

### Sintaxis utilizada

La instrucción `!pip install -q xlrd` ejecuta un comando del sistema desde una celda de Colab e instala la biblioteca `xlrd`, utilizada para leer archivos Excel antiguos con extensión `.xls`.

Luego, `pd.read_excel(ruta_xls)` abre el archivo indicado y lo convierte en un DataFrame, que almacenamos en `df_xls`.

Finalmente, `df_xls.head()` muestra las primeras cinco filas del dataset.


## Resumen

Aunque los cuatro archivos utilizan formatos distintos, en todos los casos el objetivo es el mismo: convertir los datos en un **DataFrame de Pandas**.

Una vez realizada la carga, podemos trabajar con `df_txt`, `df_csv`, `df_json` y `df_xls` utilizando las mismas herramientas de selección, filtrado, análisis y visualización de Pandas.
