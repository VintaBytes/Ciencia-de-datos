### Opciones del parámetro `keep`

El método `duplicated()` permite indicar qué ocurrencia de un registro repetido debe tomarse como referencia mediante el parámetro `keep`.

Las opciones principales son:

```python
keep="first"
```

Conserva la primera aparición como referencia y marca como duplicadas las siguientes. Es el comportamiento por defecto.

```python
keep="last"
```

Conserva la última aparición como referencia y marca como duplicadas las anteriores.

```python
keep=False
```

Marca como duplicadas **todas** las filas que pertenecen a un grupo repetido, incluida la primera aparición.

Para inspeccionar los registros duplicados utilizaremos `keep=False`, ya que queremos ver todas las filas involucradas.


````python
df[df.duplicated(keep=False)]
````

## ¿Puede repetirse el identificador de un empleado?

Eliminar filas duplicadas completas no garantiza que todos los registros sean únicos desde el punto de vista lógico.

Por ejemplo, dos filas pueden contener datos diferentes y, aun así, compartir el mismo identificador.

La columna `Employee_ID` parece cumplir la función de identificar a cada empleado. Por eso conviene comprobar si alguno de sus valores aparece más de una vez.

Para hacerlo podemos utilizar:

```python
df["Employee_ID"].duplicated().sum()
```

En este caso, `duplicated()` ya no compara filas completas: analiza solamente los valores de la columna `Employee_ID`.


````python
df["Employee_ID"].duplicated().sum()
````

Esto significa que existen **8 valores de `Employee_ID` que aparecen nuevamente después de una primera aparición**.

Es importante notar que estos casos no fueron eliminados por `drop_duplicates()`. Por lo tanto, las filas involucradas no son copias completas entre sí: comparten el mismo identificador, pero al menos algún otro dato es diferente.

Antes de decidir qué hacer con estos registros, necesitamos verlos.


````python
df[df["Employee_ID"].duplicated(keep=False)].sort_values("Employee_ID")
````

La búsqueda mostró **16 filas**, correspondientes a **8 identificadores de empleado repetidos**.

A diferencia de los duplicados completos que encontramos anteriormente, en estos casos las dos filas asociadas a un mismo `Employee_ID` **no son idénticas**.

Al compararlas podemos observar dos situaciones diferentes.

En algunos empleados cambia el valor de `Salary`. Por ejemplo, para `EMP1238` aparecen:

```text
64012.48
65932.85
```

mientras que el resto de los datos mostrados para ese empleado coincide.

En otros casos cambia `Status`. Por ejemplo, `EMP1367` aparece una vez como:

```text
Active
```

y otra como:

```text
Pending
```

La misma clase de diferencia aparece en otros identificadores repetidos.

Esto nos presenta un problema distinto al de los duplicados exactos:

> **sabemos que existen registros contradictorios, pero el propio dataset no nos permite determinar cuál de los dos valores es el correcto.**

No sería apropiado elegir automáticamente la primera o la última fila solamente por su posición en el archivo.

En un proyecto real, este tipo de situación debería resolverse consultando una fuente de información confiable, estableciendo una regla de negocio o verificando el origen de los registros.

En nuestro caso no contamos con información externa que permita decidir cuál de las dos versiones es correcta. Por lo tanto, debemos adoptar una estrategia explícita **y documentarla**.


### Tratamiento de los identificadores conflictivos

Como no podemos determinar cuál de las dos versiones de cada empleado es correcta, optaremos por eliminar del análisis **todas las filas correspondientes a identificadores repetidos**.

Esta decisión implica perder algunos registros, pero evita conservar información cuya validez no podemos establecer.

Primero identificaremos los `Employee_ID` que aparecen más de una vez y luego eliminaremos todas las filas asociadas a ellos.

### ¿Cómo eliminaremos los identificadores repetidos?

Para eliminar correctamente estos registros necesitamos realizar tres pasos.

Primero utilizaremos:

```python
df["Employee_ID"].duplicated(keep=False)
```

Con `keep=False`, Pandas marca como repetidas **todas las apariciones** de cada `Employee_ID` duplicado, no solamente la segunda.

Luego emplearemos:

```python
isin()
```

Este método permite comprobar si cada valor de `Employee_ID` pertenece al conjunto de identificadores repetidos.

Finalmente utilizaremos el operador:

```python
~
```

que representa una negación lógica.

De esta manera podemos conservar solamente las filas cuyo `Employee_ID` **no** pertenece al conjunto de identificadores conflictivos.

En resumen:

* `duplicated(keep=False)` localiza todas las apariciones conflictivas;
* `isin()` comprueba si cada ID pertenece a ese conjunto;
* `~` niega la condición, por lo que conservamos los registros cuyos IDs **no** están entre los repetidos.


````python
ids_repetidos = df.loc[
    df["Employee_ID"].duplicated(keep=False),
    "Employee_ID"
]

df = df[~df["Employee_ID"].isin(ids_repetidos)]

df.shape
````

El DataFrame quedó con:

```text
1012 filas
12 columnas
```

Antes de esta operación tenía 1028 filas.

Por lo tanto, se eliminaron **16 registros**, correspondientes a las dos apariciones de cada uno de los 8 identificadores conflictivos detectados.

Las 12 columnas se mantuvieron sin cambios.

A partir de este punto continuaremos trabajando únicamente con registros cuyo `Employee_ID` aparece una sola vez.

## Detección de outliers mediante el rango intercuartílico

Una forma habitual de detectar valores atípicos en una variable numérica consiste en utilizar los **cuartiles**.

El primer cuartil (`Q1`) deja por debajo aproximadamente al 25 % de los datos, mientras que el tercer cuartil (`Q3`) deja por debajo aproximadamente al 75 %.

La diferencia entre ambos se denomina **rango intercuartílico**:

$$
IQR = Q3 - Q1
$$

Un criterio frecuente considera potenciales outliers a los valores que se encuentran fuera del intervalo:

$$
Q1 - 1.5 \times IQR
$$

y

$$
Q3 + 1.5 \times IQR
$$

Este método no determina que un dato sea necesariamente incorrecto. Solamente permite identificar valores suficientemente alejados de la zona central de la distribución como para justificar una inspección adicional.

En Pandas podemos obtener los cuartiles mediante:

```python
df["Salary"].quantile(0.25)
df["Salary"].quantile(0.75)
```

Calcularemos `Q1`, `Q3`, el rango intercuartílico y los límites inferior y superior resultantes.


````python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
````

Obtuvimos:

```text
Q1 = 68254.95
Q3 = 101245.38
IQR = 32990.43
```

Aplicando el criterio de `1.5 × IQR`, los límites resultaron aproximadamente:

```text
Límite inferior = 18769.30
Límite superior = 150731.03
```

Según este criterio, cualquier salario menor que `18769.30` o mayor que `150731.03` puede considerarse un **valor atípico potencial**.

> Esto no significa automáticamente que esos registros sean erróneos. El método solamente nos indica que se encuentran suficientemente alejados de la zona central de la distribución como para justificar una revisión.

El siguiente paso será identificar exactamente qué salarios quedan fuera de esos límites.


### Selección de salarios fuera de los límites

Para localizar los posibles outliers podemos construir una condición lógica:

```python
(df["Salary"] < limite_inferior) | (df["Salary"] > limite_superior)
```

El símbolo:

```python
|
```

representa una operación lógica **OR**: la condición se cumple si el salario está por debajo del límite inferior **o** por encima del límite superior.

Utilizaremos esa condición para seleccionar solamente las filas que cumplen alguno de esos dos casos.


````python
outliers_salary = df[
    (df["Salary"] < limite_inferior) |
    (df["Salary"] > limite_superior)
]

outliers_salary[["Employee_ID", "Salary"]].sort_values("Salary")
````

Aplicando los límites calculados mediante el rango intercuartílico, encontramos **10 registros** cuyos salarios quedan fuera del intervalo considerado habitual.

Los cinco primeros se encuentran por debajo del límite inferior calculado, mientras que los cinco últimos superan el límite superior.

En este caso, además, algunos valores resultan claramente problemáticos desde el punto de vista semántico: por ejemplo, un salario negativo o igual a cero difícilmente represente un salario válido.

En cambio, los valores muy altos requieren algo más de cautela. Un valor extremo puede ser perfectamente real en determinados contextos.

Por eso conviene distinguir entre dos conceptos:

> **outlier** no significa necesariamente **error**.

El criterio IQR nos ayuda a localizar observaciones que merecen revisión, pero la decisión final debería considerar también el significado de la variable y el contexto del problema.

Para este dataset didáctico adoptaremos una regla práctica: los valores detectados fuera de los límites del IQR serán tratados como valores inválidos y reemplazados por `NaN`.

De esta manera podremos gestionar posteriormente todos los salarios faltantes o inválidos mediante una estrategia común.


### Reemplazo de los salarios atípicos

Utilizaremos la misma condición lógica que empleamos para detectarlos:

```python
(df["Salary"] < limite_inferior) | (df["Salary"] > limite_superior)
```

Con `loc[]` seleccionaremos solamente esas filas y la columna `Salary`. Luego asignaremos:

```python
pd.NA
```

para indicar que esos valores deben considerarse ausentes o no válidos.


````python
df.loc[
    (df["Salary"] < limite_inferior) |
    (df["Salary"] > limite_superior),
    "Salary"
] = pd.NA

df["Salary"].isna().sum()
````

Después de reemplazar por `NaN` los salarios que quedaban fuera de los límites definidos mediante el criterio IQR, la columna `Salary` quedó con 34 valores ausentes.

Antes de esta operación habíamos observado que `Salary` tenía 24 valores ausentes. Por lo tanto, **se agregaron 10 nuevos valores ausentes**.


Estos corresponden exactamente a los 10 salarios que habíamos identificado como outliers mediante el criterio utilizado.

De esta manera, los valores faltantes originales y los valores considerados inválidos quedaron representados de una forma común mediante `NaN`.

Más adelante tendremos que decidir qué estrategia utilizar para tratar esos valores ausentes. Por ahora continuaremos explorando otras columnas del dataset antes de realizar imputaciones o eliminaciones definitivas.

### Conversión de `Remote_Work` a valores booleanos

```text
true, yes, y, 1  → True
```

Finalmente convertiremos la columna al tipo:

```python

```

Este tipo de Pandas permite almacenar tres posibilidades:

## Revisión de `Email`

La columna `Email` contiene direcciones de correo electrónico.

En este caso no buscamos convertir el tipo de dato, sino detectar valores que puedan tener errores de formato.

Una validación sencilla consiste en comprobar si cada dirección contiene una estructura mínima del tipo:

```text
texto@texto.dominio
```

No será una validación exhaustiva de todos los correos posibles, pero sí nos permitirá localizar errores evidentes.

Utilizaremos `str.match()` con una expresión regular sencilla.

La expresión:

```python
^[^@\s]+@[^@\s]+\.[^@\s]+$
```

comprueba, de manera básica, que exista:

* texto antes de `@`;
* un separador `@`;
* texto después de `@`;
* un punto antes del dominio final;
* ausencia de espacios.

Además utilizaremos:

```python
na=False
```

para evitar que los valores ausentes generen problemas durante la comparación.


````python
patron_email = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

emails_invalidos = df[
    df["Email"].notna() &
    ~df["Email"].str.match(patron_email, na=False)
]

emails_invalidos[["Employee_ID", "Email"]]
````

La validación encontró varios valores claramente problemáticos, por ejemplo:

```text
mary@@example.com
@example.com
no_email
lisa.white@ example.com
grace.brownexample.com
john doe@example.com
bob.smith@
eva.miller@example
```

Sin embargo, también aparecen algunos correos que visualmente parecen correctos, como:

```text
carol.davis@example.com
dave.wilson@example.com
```

Esto sugiere que podrían contener espacios u otros caracteres poco visibles. Antes de corregir la columna conviene inspeccionar esos valores utilizando nuevamente `repr()`.


````python
for valor in emails_invalidos["Email"]:
    print(repr(valor))
````

Algunos valores solamente tienen espacios sobrantes:

```text
' carol.davis@example.com'
'dave.wilson@example.com '
```

Estos casos pueden corregirse fácilmente eliminando los espacios al comienzo y al final de la cadena. En cambio, otros valores presentan errores de estructura, por ejemplo:

```text
'mary@@example.com'
'@example.com'
'no_email'
'lisa.white@ example.com'
'grace.brownexample.com'
'john doe@example.com'
'bob.smith@'
'eva.miller@example'
```

Como no podemos reconstruir con certeza la dirección correcta en esos casos, los trataremos como valores ausentes. Primero eliminaremos espacios al comienzo y al final con:

```python
.str.strip()
```

Luego volveremos a aplicar el patrón de validación y convertiremos en `pd.NA` aquellos valores que sigan sin cumplirlo.


````python
df["Email"] = df["Email"].str.strip()

email_valido = df["Email"].str.match(patron_email, na=False)

df.loc[
    df["Email"].notna() & ~email_valido,
    "Email"
] = pd.NA

df["Email"].isna().sum()
````

La columna `Email` quedó con 14 valores ausentes. Los correos que solo tenían espacios al comienzo o al final pudieron conservarse después de la normalización.

En cambio, los valores con errores de estructura quedaron representados como ausentes, ya que no disponemos de información suficiente para reconstruirlos correctamente.

## Tratamiento de valores ausentes

Primero revisaremos nuevamente cuántos valores ausentes quedan en cada columna.

## ¿Qué proporción de datos falta en cada columna?

El número absoluto de valores ausentes es útil, pero no siempre alcanza para evaluar su importancia.

Por ejemplo, 20 valores faltantes pueden ser pocos en un dataset de 100.000 registros, pero representar una proporción importante en uno de 100 filas.

Por eso calcularemos también el **porcentaje de valores ausentes** de cada columna.

Para ello:

* `isna().sum()` cuenta los valores faltantes;
* `len(df)` indica la cantidad total de filas;
* multiplicamos por `100` para expresar el resultado como porcentaje;
* `round(2)` limita la visualización a dos decimales.


````python
faltantes = pd.DataFrame({
    "Cantidad": df.isna().sum(),
    "Porcentaje": (df.isna().sum() / len(df) * 100).round(2)
})

faltantes[faltantes["Cantidad"] > 0]
````

## Estrategias para tratar los valores ausentes

Los porcentajes muestran situaciones diferentes.

La columna `Age` tiene un **21,15 %** de valores ausentes, por lo que eliminar todas esas filas supondría descartar una parte importante del dataset.

En `Salary`, en cambio, los faltantes representan un **3,36 %**, y en las demás variables el porcentaje es todavía menor.

Una alternativa a eliminar registros consiste en **imputar** los valores faltantes: reemplazarlos utilizando algún criterio razonable.

Para variables numéricas son frecuentes estrategias como:

* reemplazar por la **media**;
* reemplazar por la **mediana**;
* utilizar algún valor calculado a partir de otras variables.

La elección no debería hacerse automáticamente. Antes conviene conocer las medidas de tendencia central de nuestras variables numéricas.

Calcularemos la media, la mediana y la moda de `Age` y `Salary` antes de decidir qué utilizar.


````python
print("Age")
print("Media:  ", df["Age"].mean())
print("Mediana:", df["Age"].median())
print("Moda:   ", df["Age"].mode().tolist())

print("\nSalary")
print("Media:  ", df["Salary"].mean())
print("Mediana:", df["Salary"].median())
print("Moda:   ", df["Salary"].mode().tolist())
````

## Elección de una estrategia de imputación

Las medidas calculadas muestran situaciones diferentes.

Para `Age` obtuvimos:

```text
Media   = 32.52
Mediana = 30
Moda    = 40
```

La variable solo contiene actualmente las edades `25`, `30`, `35` y `40`.

Si utilizáramos la media para completar los valores faltantes introduciríamos edades como `32.52`, que no aparecen entre los valores observados y que, en este dataset, resultarían poco naturales.

La mediana, en cambio, es:

```text
30
```

y corresponde a uno de los valores existentes en la columna. Por eso utilizaremos la **mediana** para imputar `Age`.

Para `Salary` encontramos:

```text
Media   ≈ 85221.22
Mediana ≈ 85621.85
```

Ambos valores son bastante próximos.

Además, la salida de `mode()` contiene muchos resultados, por lo que no existe una única moda claramente representativa.

Como ya eliminamos los valores extremos detectados y la mediana es una medida robusta frente a posibles valores alejados del centro, utilizaremos también la **mediana** para completar los salarios faltantes.

> La estrategia de imputación elegida forma parte de las decisiones de preparación del dataset y debe quedar documentada.


### Imputación de `Age` y `Salary`

El método `fillna()` permite reemplazar valores ausentes.

Calcularemos la mediana de cada variable y utilizaremos ese valor para completar sus `NaN`.


````python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

print("Valores ausentes en Age:", df["Age"].isna().sum())
print("Valores ausentes en Salary:", df["Salary"].isna().sum())
````

## Resultado de la imputación numérica

Después de completar los valores faltantes con la mediana correspondiente:

```text
Age    → 0 valores ausentes
Salary → 0 valores ausentes
```

De esta manera, ambas variables numéricas quedaron completas sin necesidad de eliminar registros adicionales.


## Valores ausentes en variables categóricas

Para las variables categóricas, una estrategia posible consiste en reemplazar los valores ausentes por la categoría más frecuente.

Antes de hacerlo, revisaremos cuál es la moda de cada columna.


````python
columnas_categoricas = [
    "Status",
    "Department",
    "Region",
    "Performance_Score",
    "Remote_Work"
]

for columna in columnas_categoricas:
    print(columna, "==>", df[columna].mode().tolist())
````

## Imputación de variables categóricas

En las columnas analizadas obtuvimos una categoría modal clara:

```text
Status            ==> Pending
Department        ==> DevOps
Region            ==> California
Performance_Score ==> Good
Remote_Work       ==> True
```

Como la proporción de valores ausentes en estas variables es pequeña, utilizaremos la **moda** de cada columna para completar los datos faltantes.

Esta es una estrategia sencilla y frecuente para variables categóricas, aunque no debe aplicarse automáticamente en todos los problemas: siempre conviene considerar el significado de la variable y el contexto de los datos.


````python
columnas_categoricas = [
    "Status",
    "Department",
    "Region",
    "Performance_Score",
    "Remote_Work"
]

for columna in columnas_categoricas:
    df[columna] = df[columna].fillna(df[columna].mode()[0])

df[columnas_categoricas].isna().sum()
````

Después de completar cada variable categórica con su moda, todas las columnas tratadas quedaron sin valores ausentes. Con esto quedan completas tanto las variables numéricas `Age` y `Salary` como las variables categóricas que acabamos de tratar.

para indicar explícitamente qué columnas queremos quitar.


````python
columnas_a_eliminar = [
    "Employee_ID",
    "First_Name",
    "Last_Name",
    "Email",
    "Phone",
    "Department_Region",
    "Join_Date"
]

df_ml = df.drop(columns=columnas_a_eliminar)

df_ml.head()
````

```text
Age
```

Las columnas utilizadas principalmente como identificadores, datos personales o información redundante quedaron fuera de esta nueva versión.

### Revisión de los tipos de datos de `df_ml`

Antes de transformar las variables categóricas, comprobaremos qué tipo de dato tiene actualmente cada columna.


````python
df_ml.dtypes
````

La revisión de `df_ml.dtypes` muestra que las variables categóricas que todavía necesitan transformación son:

````python
df_ml["Join_Year"] = df_ml["Join_Year"].astype("int64")
df_ml["Join_Month"] = df_ml["Join_Month"].astype("int64")

df_ml[["Join_Year", "Join_Month"]].dtypes
````

## Variables categóricas: nominales y ordinales

No todas las variables categóricas tienen la misma naturaleza.

En `df_ml` encontramos, por ejemplo:

```text
Status
Department
Region
Performance_Score
```

Las variables `Status`, `Department` y `Region` son **nominales**: sus categorías representan grupos diferentes, pero no existe un orden natural entre ellas.

Por ejemplo:

```text
California
Florida
Texas
```

son simplemente categorías distintas.

En cambio, `Performance_Score` es una variable **ordinal**, porque sus categorías sí poseen un orden:

```text
Poor < Average < Good < Excellent
```

Esta diferencia es importante porque no conviene codificar ambos tipos de variables de la misma manera.

Para las variables nominales utilizaremos más adelante una codificación que no introduzca un orden artificial.

Para `Performance_Score`, en cambio, podemos asignar valores numéricos respetando su jerarquía.


### Codificación ordinal de `Performance_Score`

Asignaremos los siguientes valores:

```text
Poor      ==> 0
Average   ==> 1
Good      ==> 2
Excellent ==> 3
```

De esta forma, los números conservan el orden conceptual de las categorías.


````python
orden_performance = {
    "Poor": 0,
    "Average": 1,
    "Good": 2,
    "Excellent": 3
}

df_ml["Performance_Score"] = df_ml["Performance_Score"].map(orden_performance)

df_ml["Performance_Score"].value_counts().sort_index()
````

La variable `Performance_Score` quedó transformada y la variable conserva el orden natural de las categorías y ya puede ser utilizada como una característica numérica.


## Codificación de variables nominales con scikit-learn

Las variables:

```text
Status
Department
Region
```

son categóricas nominales, es decir, sus categorías no tienen un orden natural.

Para transformarlas utilizaremos `OneHotEncoder` de **scikit-learn**.

Este transformador crea una nueva columna binaria por cada categoría encontrada.

Por ejemplo, una variable como:

```text
Status
```

puede convertirse en columnas como:

```text
Status_Active
Status_Inactive
Status_Pending
```

En cada fila, una de esas columnas tendrá valor `1` y las demás `0`, según la categoría correspondiente.

A diferencia de `pd.get_dummies()`, `OneHotEncoder` forma parte del ecosistema de scikit-learn y más adelante podremos incorporarlo directamente dentro de un `Pipeline`.

### Configuración de `OneHotEncoder`

Utilizaremos:

```python
handle_unknown="ignore"
```

para evitar errores si en el futuro aparecen categorías que no estaban presentes durante el ajuste del codificador.

También indicaremos:

```python
sparse_output=False
```

para obtener una matriz densa, más cómoda de visualizar en este cuaderno.

Primero ajustaremos el codificador a las columnas categóricas y luego observaremos qué nombres tendrán las nuevas variables.


````python
from sklearn.preprocessing import OneHotEncoder

columnas_nominales = [
    "Status",
    "Department",
    "Region"
]

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoder.fit(df_ml[columnas_nominales])

encoder.get_feature_names_out(columnas_nominales)
````

## Variables generadas por `OneHotEncoder`

El codificador detectó las categorías presentes en `Status`, `Department` y `Region`, y preparó las siguientes variables:

```text
Status_Active
Status_Inactive
Status_Pending

Department_Admin
Department_Cloud Tech
Department_DevOps
Department_Finance
Department_HR
Department_Sales

Region_California
Region_Florida
Region_Illinois
Region_Nevada
Region_New York
Region_Texas
```

Cada una representará una categoría mediante valores binarios.

Por ejemplo, para un empleado cuyo estado sea `Active`:

```text
Status_Active   = 1
Status_Inactive = 0
Status_Pending  = 0
```

Ahora que conocemos las columnas que se crearán, podemos realizar la transformación y convertir el resultado nuevamente en un DataFrame de Pandas.

### Aplicación de la codificación

El método:

```python
encoder.transform(...)
```

aplica la transformación aprendida por `OneHotEncoder`.

El resultado es una matriz de NumPy, por lo que la convertiremos a un DataFrame utilizando los nombres obtenidos con `get_feature_names_out()`.

Finalmente eliminaremos las tres columnas categóricas originales y agregaremos las nuevas columnas codificadas.


````python
datos_codificados = encoder.transform(df_ml[columnas_nominales])

df_codificado = pd.DataFrame(
    datos_codificados,
    columns=encoder.get_feature_names_out(columnas_nominales),
    index=df_ml.index
)

df_ml = pd.concat(
    [
        df_ml.drop(columns=columnas_nominales),
        df_codificado
    ],
    axis=1
)

df_ml.head()
````

Después de aplicar `OneHotEncoder`, las columnas categóricas nominales originales fueron reemplazadas por variables binarias.

Por ejemplo, `Status` dejó de aparecer como una única columna de texto y fue reemplazada por:

```text
Status_Active
Status_Inactive
Status_Pending
```

Lo mismo ocurrió con `Department` y `Region`.

Cada nueva columna toma valores `0` o `1` según corresponda a la categoría de cada registro.

El DataFrame `df_ml` quedó ahora con **21 columnas**.

En esta etapa ya no tenemos variables nominales representadas como texto.
