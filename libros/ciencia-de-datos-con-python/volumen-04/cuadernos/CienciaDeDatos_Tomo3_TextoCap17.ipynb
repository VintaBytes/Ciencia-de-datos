 Capítulo 17 · Sobreajuste y subajuste

## Qué significa generalizar

En Machine Learning, entrenamos modelos usando datos disponibles. Esos datos contienen ejemplos que ya conocemos: clientes históricos, viviendas ya vendidas, transacciones pasadas, registros de ventas, observaciones anteriores. Durante el entrenamiento, el modelo busca patrones en esos ejemplos para poder realizar predicciones.

Pero el objetivo final no es que el modelo funcione bien únicamente sobre los datos que ya vio. El objetivo es que pueda comportarse razonablemente bien con **datos nuevos**.

A esa capacidad la llamamos **generalización**.

```text id="bjp122"
Generalizar significa:
aprender patrones útiles en los datos de entrenamiento
y aplicarlos correctamente a casos nuevos.
```

Esta idea está en el centro del Machine Learning. Un modelo no se entrena para memorizar una tabla. Se entrena para encontrar regularidades que puedan servir más allá de los ejemplos conocidos.

Por ejemplo, si entrenamos un modelo para predecir el precio de viviendas, no nos interesa solamente que estime bien los precios de las viviendas que ya estaban en el dataset. Queremos que pueda estimar el precio de una vivienda nueva, que todavía no fue usada durante el entrenamiento.

```text id="qv8myd"
Datos de entrenamiento:
viviendas ya conocidas.

Caso nuevo:
vivienda que el modelo no vio antes.

Objetivo:
producir una estimación razonable para el caso nuevo.
```

Lo mismo ocurre en clasificación. Si entrenamos un modelo para detectar fraude, no alcanza con que clasifique bien transacciones históricas. Queremos que ayude a detectar fraudes futuros. Si entrenamos un modelo para predecir abandono de clientes, queremos que identifique clientes actuales en riesgo, no solo que explique correctamente abandonos pasados.

Por eso separamos datos de entrenamiento y prueba. El conjunto de entrenamiento sirve para ajustar el modelo. El conjunto de prueba sirve para simular, de manera controlada, qué ocurre cuando el modelo se enfrenta a casos que no vio durante el entrenamiento.

```text id="rvdsfx"
Entrenamiento:
el modelo aprende.

Prueba:
evaluamos si ese aprendizaje funciona
en datos no vistos.
```

Si un modelo obtiene buen desempeño en entrenamiento y también buen desempeño en prueba, tenemos una señal positiva. No garantiza que funcionará perfectamente en el futuro, pero indica que los patrones aprendidos no dependen solo de memorizar los datos de entrenamiento.

En cambio, si un modelo funciona muy bien en entrenamiento pero mal en prueba, hay un problema. El modelo aprendió demasiado los detalles específicos de los datos conocidos, pero no logró capturar patrones que se mantengan en casos nuevos. Este fenómeno se llama sobreajuste, y lo estudiaremos en la próxima sección.

También puede ocurrir lo contrario: que el modelo funcione mal tanto en entrenamiento como en prueba. En ese caso, probablemente no está aprendiendo lo suficiente. Tal vez el modelo es demasiado simple, las variables no contienen señal útil, el problema está mal formulado o los datos no alcanzan. Este fenómeno se relaciona con el subajuste.

La generalización se ubica entre esos dos extremos. Queremos un modelo que aprenda lo suficiente como para capturar patrones reales, pero no tanto como para memorizar detalles accidentales.

```text id="wk7cpd"
Aprender demasiado poco:
subajuste.

Aprender detalles específicos del entrenamiento:
sobreajuste.

Aprender patrones que funcionan en datos nuevos:
generalización.
```

Una forma intuitiva de pensar esto es comparar un modelo con un estudiante. Si un estudiante memoriza las respuestas exactas de una guía, puede obtener buen resultado si el examen repite las mismas preguntas. Pero si el examen cambia los datos o plantea situaciones nuevas, probablemente tendrá dificultades. No aprendió el concepto; memorizó ejemplos.

Un modelo sobreajustado se parece a ese estudiante. Puede funcionar muy bien con los datos conocidos, pero falla cuando aparecen casos nuevos.

En cambio, un estudiante que entiende el procedimiento puede resolver ejercicios diferentes, aunque no sean idénticos a los practicados. No necesita que el problema sea exactamente igual. Reconoce la estructura general y la aplica a una situación nueva. Eso se parece más a generalizar.

```text id="jn5xuh"
Memorizar ejemplos:
puede funcionar en entrenamiento.

Comprender patrones:
permite responder mejor ante casos nuevos.
```

En Machine Learning, esta diferencia es fundamental porque los datos de entrenamiento siempre son una muestra limitada de la realidad. No contienen todos los casos posibles. Pueden tener ruido, errores, valores extremos, ausencias, sesgos o situaciones particulares del período en que fueron recolectados. Si el modelo se ajusta demasiado a esos detalles, puede perder capacidad para funcionar fuera de esa muestra.

Generalizar no significa acertar siempre. Ningún modelo real predice perfectamente todos los casos nuevos. Generalizar significa que el modelo mantiene un desempeño razonable cuando se enfrenta a datos que no fueron usados para entrenarlo.

Por ejemplo, un modelo de regresión puede tener cierto error promedio en entrenamiento y un error parecido en prueba. Eso sugiere que el modelo está aprendiendo una relación que se mantiene en datos no vistos. Si el error de prueba es mucho mayor, la señal es distinta: el modelo aprendió algo que no se sostiene fuera del entrenamiento.

```text id="mygyrv"
Buen indicio:
error de entrenamiento y error de prueba
razonablemente cercanos.

Señal de alerta:
error de entrenamiento muy bajo
y error de prueba mucho más alto.
```

En clasificación ocurre algo similar. Si la accuracy, el F1-score o las métricas relevantes son muy altas en entrenamiento pero caen mucho en prueba, debemos sospechar. El modelo puede estar adaptándose demasiado a los casos conocidos.

Por eso, la evaluación sobre datos no vistos no es un detalle técnico. Es la forma básica de comprobar si el modelo generaliza.

También debemos entender que generalizar depende del parecido entre los datos de entrenamiento y los datos futuros. Si entrenamos un modelo con datos de una ciudad y luego lo usamos en otra con características muy distintas, el desempeño puede cambiar. Si entrenamos con datos de un período económico estable y luego lo usamos durante una crisis, las relaciones aprendidas pueden dejar de funcionar. Si entrenamos con un tipo de usuario y luego aplicamos el modelo a otro grupo, puede fallar.

```text id="g729dm"
Un modelo generaliza mejor cuando los datos nuevos
se parecen razonablemente a los datos usados para entrenar.
```

Esto no significa que los casos nuevos deban ser idénticos. Si fueran idénticos, no necesitaríamos generalizar. Pero sí debe existir cierta continuidad entre el pasado que usamos para entrenar y el futuro o contexto donde aplicaremos el modelo.

Esta idea ya había aparecido cuando hablamos de variables disponibles antes de la predicción y de fuga de datos. Un modelo puede parecer excelente si se entrena con información que no estará disponible en el momento real de uso. Pero ese desempeño no representa verdadera generalización. Representa una evaluación engañosa.

Generalizar también implica evitar aprender ruido. En los datos puede haber patrones accidentales que no se repetirán. Por ejemplo, una promoción puntual puede haber aumentado ventas en una semana específica. Un error de carga puede hacer que ciertos valores parezcan relacionados. Un grupo pequeño de casos puede tener una coincidencia casual. Si el modelo toma esas señales como si fueran reglas generales, puede equivocarse en el futuro.

```text id="bg3vnm"
No todo patrón en los datos de entrenamiento
es un patrón útil para datos nuevos.
```

Por eso, un modelo más complejo no siempre generaliza mejor. Un modelo muy flexible puede adaptarse a muchos detalles del entrenamiento, incluso a los que no representan relaciones reales. En cambio, un modelo más simple puede ignorar parte del ruido y capturar una tendencia más estable. La clave está en encontrar un equilibrio.

Este equilibrio será el tema central del capítulo. El sobreajuste aparece cuando el modelo se adapta demasiado a los datos de entrenamiento. El subajuste aparece cuando el modelo es tan simple o insuficiente que no captura ni siquiera los patrones principales. La buena generalización está entre ambos.

```text id="f3vbxf"
Queremos un modelo que aprenda patrones,
no un modelo que memorice ruido
ni uno que ignore la señal disponible.
```

En la práctica, para evaluar generalización comparamos el desempeño en entrenamiento y en prueba. Más adelante, usaremos validación cruzada para tener una evaluación más estable. Pero la intuición ya está presente: un modelo útil debe funcionar en datos que no vio.

> Generalizar significa que un modelo aprende relaciones a partir de los datos de entrenamiento y logra aplicarlas razonablemente bien a casos nuevos. Es la diferencia entre memorizar ejemplos y capturar patrones útiles. La separación entre entrenamiento y prueba existe justamente para medir esa capacidad. Cuando un modelo no generaliza bien, puede estar sobreajustado, subajustado o afectado por problemas en los datos, las variables o la forma de evaluación.


## Overfitting

El **overfitting**, o **sobreajuste**, ocurre cuando un modelo aprende demasiado bien los datos de entrenamiento, al punto de capturar no solo patrones útiles, sino también detalles accidentales, ruido, valores particulares o irregularidades que no se repetirán en datos nuevos.

Un modelo sobreajustado puede parecer excelente durante el entrenamiento. Sus métricas sobre los datos conocidos pueden ser muy buenas: pocos errores, alta accuracy, bajo MAE, alto R². Pero cuando lo evaluamos con datos de prueba, su rendimiento cae. Esa diferencia revela que el modelo no aprendió una regla general, sino una versión demasiado ajustada a los ejemplos que vio.

```text id="jyec3m"
Sobreajuste:
muy buen desempeño en entrenamiento,
pero mal desempeño en datos no vistos.
```

La idea puede parecer contradictoria al principio. ¿Cómo puede ser malo que un modelo aprenda mucho? El problema es que no todo lo que aparece en los datos de entrenamiento representa una regularidad útil. Los datos reales contienen ruido. Pueden incluir errores de carga, casos atípicos, coincidencias casuales, pequeñas variaciones sin importancia o patrones que solo existen en esa muestra.

Si el modelo tiene suficiente flexibilidad, puede adaptarse a esos detalles. En lugar de aprender la tendencia general, empieza a aprender particularidades del conjunto de entrenamiento.

Por ejemplo, imaginemos un modelo de regresión que predice precios de viviendas. Una relación general útil podría ser que, en promedio, viviendas con más superficie tienden a tener precios más altos. Pero el dataset puede contener casos particulares: una vivienda pequeña muy cara por su ubicación, una vivienda grande barata por su estado, una operación registrada con error o una propiedad excepcional que no representa al mercado general.

Un modelo demasiado flexible podría intentar ajustar todos esos casos particulares. En entrenamiento, sus errores serían muy bajos. Pero al recibir nuevas viviendas, esas reglas tan específicas probablemente no se mantendrán.

```text id="y6uafl"
Patrón útil:
a mayor superficie, mayor precio estimado, en promedio.

Detalle accidental:
esta vivienda particular se comporta distinto
por una razón que el modelo no puede generalizar.
```

En clasificación ocurre algo similar. Supongamos un modelo para detectar abandono de clientes. Puede aprender patrones razonables, como baja frecuencia de uso, muchos reclamos recientes o caída en la interacción. Pero también podría aprender coincidencias demasiado específicas: combinaciones raras de variables que aparecen en pocos clientes del entrenamiento y que no representan un patrón estable.

El modelo puede clasificar perfectamente a esos clientes conocidos, pero fallar con clientes nuevos.

Una forma intuitiva de entender el sobreajuste es volver a la analogía del estudiante. Un estudiante que memoriza las respuestas exactas de una guía puede obtener un resultado perfecto si el examen repite los mismos ejercicios. Pero si el examen cambia los valores o plantea una situación parecida con otra redacción, se equivoca. No aprendió el concepto; memorizó los ejemplos.

Un modelo sobreajustado se comporta de manera parecida. Aprende demasiado los ejemplos de entrenamiento y pierde capacidad para resolver casos nuevos.

```text id="n7qoh2"
Memorizar entrenamiento:
buen resultado sobre datos conocidos.

Aprender patrones generales:
mejor resultado sobre datos nuevos.
```

El sobreajuste suele aparecer cuando el modelo es demasiado complejo para la cantidad o calidad de datos disponibles. Un árbol de decisión muy profundo, por ejemplo, puede crear muchas divisiones hasta separar casi perfectamente los casos de entrenamiento. Puede terminar con hojas que contienen muy pocos ejemplos, incluso uno solo. Eso le permite memorizar el dataset, pero no necesariamente generalizar.

En regresión, un modelo demasiado flexible puede seguir cada ondulación de los datos, incluso las que corresponden a ruido. En lugar de capturar una tendencia general, se ajusta a cada punto conocido. Visualmente, sería como dibujar una línea que pasa por casi todos los puntos, pero con una forma tan irregular que no representa bien el fenómeno.

```text id="z5f01x"
Modelo demasiado flexible:
puede adaptarse al ruido.

Modelo que generaliza:
captura la tendencia estable.
```

El sobreajuste no depende solo del algoritmo. También depende de la cantidad de datos, la cantidad de variables, la presencia de ruido, la calidad de las etiquetas, la forma de preprocesar y la dificultad del problema. Un modelo puede sobreajustar porque es muy complejo, pero también porque tiene demasiadas variables irrelevantes o porque el conjunto de entrenamiento es pequeño.

Si tenemos pocas filas y muchas columnas, el modelo puede encontrar relaciones accidentales. Algunas variables parecerán útiles simplemente por coincidencia. En entrenamiento, esas coincidencias pueden mejorar las métricas. En datos nuevos, desaparecen.

```text id="m5zmpk"
Pocos datos + muchas variables
→ mayor riesgo de aprender coincidencias accidentales.
```

También puede haber sobreajuste cuando probamos muchas variantes del modelo y elegimos la que mejor funciona en el conjunto de prueba. Aunque parezca que estamos evaluando correctamente, si usamos el test una y otra vez para tomar decisiones, podemos terminar ajustándonos indirectamente a ese conjunto. Este punto se entenderá mejor cuando estudiemos validación cruzada y búsqueda de hiperparámetros.

Por ahora, la señal más importante es comparar entrenamiento y prueba. Si el desempeño en entrenamiento es mucho mejor que en prueba, debemos sospechar sobreajuste.

En clasificación, podríamos ver algo como:

```text id="xndesr"
Accuracy en entrenamiento: 0.99
Accuracy en prueba:        0.72
```

O en regresión:

```text id="yr37vz"
MAE en entrenamiento:  2.1
MAE en prueba:        18.7
```

En ambos casos, el modelo parece haber aprendido muy bien los datos de entrenamiento, pero no logra trasladar ese desempeño a datos no vistos.

Esto no significa que las métricas de entrenamiento deban ser idénticas a las de prueba. Es normal que el modelo funcione algo mejor en entrenamiento, porque esos datos fueron usados para ajustarlo. Lo preocupante es que la diferencia sea grande.

```text id="q2wgh1"
Diferencia pequeña:
puede ser normal.

Diferencia grande:
posible sobreajuste.
```

El sobreajuste también puede producir una falsa sensación de éxito. Si solo miramos las métricas de entrenamiento, podríamos creer que el modelo es excelente. Por eso insistimos tanto en separar datos de prueba y evaluar sobre casos no vistos. La evaluación en entrenamiento no alcanza para saber si un modelo aprendió algo útil.

Un caso especialmente engañoso aparece cuando el modelo obtiene resultados casi perfectos. En problemas reales, una accuracy de 1.00, un F1-score perfecto o un error prácticamente cero pueden ser señales de que algo anda mal. No siempre, pero conviene revisar. Puede haber sobreajuste, fuga de datos, evaluación sobre entrenamiento o una variable que revela directamente la respuesta.

```text id="yuhvye"
Resultados demasiado buenos para ser ciertos
merecen revisión.
```

Para reducir el sobreajuste existen varias estrategias. Algunas dependen del modelo. En árboles de decisión, por ejemplo, podemos limitar la profundidad, exigir más casos por hoja o evitar divisiones demasiado específicas. En otros modelos, podemos usar regularización, ajustar hiperparámetros o elegir una estructura más simple.

También podemos mejorar los datos: conseguir más ejemplos, eliminar variables irrelevantes, corregir errores, revisar outliers, evitar fuga de datos o construir variables más estables. Muchas veces, mejorar la calidad de los datos ayuda más que cambiar de algoritmo.

```text id="ax6kcj"
Para reducir sobreajuste podemos:
simplificar el modelo,
limitar su flexibilidad,
mejorar los datos,
revisar variables,
evaluar con más cuidado.
```

Sin embargo, no debemos interpretar el sobreajuste como algo que se elimina por completo. Siempre existe cierta diferencia entre aprender de una muestra y aplicar el modelo a nuevos casos. El objetivo es controlar ese riesgo y elegir modelos que mantengan un desempeño razonable fuera del entrenamiento.

También debemos evitar la reacción opuesta: simplificar tanto el modelo que ya no aprenda lo suficiente. Si reducimos demasiado la complejidad, podemos caer en subajuste. Por eso, el problema no es que un modelo aprenda mucho, sino que aprenda detalles que no generalizan.

El equilibrio consiste en permitir que el modelo capture patrones reales sin perseguir cada ruido del entrenamiento.

```text id="qptu3v"
Demasiada flexibilidad:
riesgo de sobreajuste.

Demasiada simplicidad:
riesgo de subajuste.

Buen modelo:
aprende patrones útiles y generalizables.
```

> El overfitting o sobreajuste ocurre cuando un modelo se adapta demasiado a los datos de entrenamiento y pierde capacidad para funcionar bien con datos nuevos. Suele verse como una gran diferencia entre el desempeño en entrenamiento y en prueba. Puede deberse a modelos demasiado complejos, pocos datos, ruido, variables irrelevantes o evaluación inadecuada. Detectarlo y controlarlo es fundamental para construir modelos que realmente generalicen.


## Underfitting

El **underfitting**, o **subajuste**, ocurre cuando un modelo es demasiado simple o insuficiente para capturar los patrones importantes de los datos. A diferencia del sobreajuste, donde el modelo aprende demasiado los detalles del entrenamiento, en el subajuste el modelo aprende demasiado poco.

Un modelo subajustado suele funcionar mal tanto en entrenamiento como en prueba. No logra buenos resultados ni siquiera sobre los datos que vio durante el ajuste. Esto indica que no está capturando adecuadamente la relación entre las variables de entrada y la variable objetivo.

```text id="r7c2dw"
Subajuste:
mal desempeño en entrenamiento
y mal desempeño en datos de prueba.
```

Esta es una diferencia importante con el sobreajuste. En el sobreajuste, el modelo parece excelente en entrenamiento pero falla en prueba. En el subajuste, el problema aparece desde el principio: el modelo no logra explicar bien los datos conocidos.

Por ejemplo, imaginemos que queremos predecir precios de viviendas usando solo la superficie. Si el precio depende mucho de la ubicación, el estado, la antigüedad, la cantidad de ambientes y otros factores, una regresión lineal simple con una sola variable puede quedarse corta. Tal vez capture una tendencia general, pero sus errores serán grandes porque le falta información relevante.

```text id="hkeeo8"
Modelo demasiado simple:
usa superficie solamente.

Problema real:
el precio también depende de ubicación,
estado, antigüedad y otras variables.
```

En ese caso, el modelo no está sobreajustando. No está memorizando detalles. Al contrario: no tiene suficiente capacidad o información para representar el fenómeno. Está subajustando.

El subajuste también puede aparecer cuando elegimos un modelo cuya forma no se adapta al problema. Supongamos que la relación entre una variable y el objetivo es claramente curva, pero usamos una recta. La regresión lineal intentará aproximar esa relación con una línea recta, pero tal vez falle de manera sistemática.

Por ejemplo, el consumo eléctrico puede ser bajo en temperaturas templadas y aumentar tanto en días muy fríos como en días muy calurosos. Esa relación tiene forma curva. Si usamos una regresión lineal simple con temperatura como única variable, el modelo puede no capturar bien ese comportamiento.

```text id="yj7xwv"
Relación real:
curva.

Modelo elegido:
recta.

Resultado posible:
subajuste.
```

Esto no significa que la regresión lineal sea mala. Significa que, para algunos problemas, una relación lineal básica puede ser demasiado limitada. Un buen modelo debe tener una complejidad adecuada para el fenómeno que intenta representar.

También puede haber subajuste cuando las variables de entrada no contienen suficiente señal. Si queremos predecir ventas usando columnas poco relacionadas con las ventas, el modelo tendrá dificultades aunque el algoritmo sea correcto. No puede aprender una relación que no está presente en los datos.

```text id="c6n7oz"
Si las variables no contienen señal útil,
el modelo no tiene de dónde aprender.
```

Por ejemplo, intentar predecir el precio de una vivienda usando solo un identificador interno del registro no tendría sentido. El modelo podría encontrar coincidencias accidentales, pero no una relación útil. Lo mismo ocurriría si queremos predecir abandono de clientes sin variables que describan su comportamiento, historial, uso del servicio o interacciones relevantes.

El subajuste puede deberse entonces a varias causas:

```text id="bdh5pd"
modelo demasiado simple,
pocas variables relevantes,
variables mal preparadas,
relación no lineal no capturada,
datos con poca señal,
problema mal formulado.
```

Una señal típica de subajuste es que las métricas sean malas tanto en entrenamiento como en prueba. En regresión, podríamos observar errores altos en ambos conjuntos:

```text id="m85yt5"
MAE en entrenamiento:  28.4
MAE en prueba:         30.1
```

La diferencia entre entrenamiento y prueba no es muy grande. Pero ambos errores son altos. Eso sugiere que el modelo no está aprendiendo lo suficiente.

En clasificación, podría ocurrir algo similar:

```text id="p26hej"
Accuracy en entrenamiento: 0.62
Accuracy en prueba:        0.60
```

Si el problema no es extremadamente difícil y una referencia simple obtiene valores parecidos, el modelo probablemente esté subajustando o no esté aportando información suficiente.

El subajuste también puede aparecer cuando aplicamos restricciones demasiado fuertes al modelo. Por ejemplo, un árbol de decisión con profundidad máxima muy baja puede ser incapaz de separar adecuadamente los casos. Si permitimos solo una o dos preguntas, quizá el árbol no pueda representar patrones importantes.

```text id="j78en9"
Árbol demasiado poco profundo:
puede ser demasiado simple
para capturar el problema.
```

En modelos con regularización, una regularización excesiva también puede producir subajuste. La regularización busca limitar la complejidad del modelo para evitar sobreajuste. Pero si la restricción es demasiado fuerte, el modelo puede quedar tan limitado que no aprende relaciones útiles.

Esta idea muestra que no existe una complejidad ideal universal. Un modelo demasiado complejo puede sobreajustar. Un modelo demasiado simple puede subajustar. La dificultad está en encontrar un punto intermedio.

```text id="rgbl8w"
Muy simple:
subajuste.

Muy complejo:
sobreajuste.

Complejidad adecuada:
mejor generalización.
```

Una forma intuitiva de entender el subajuste es imaginar un estudiante que estudió demasiado poco o que aprendió una regla demasiado básica para resolver problemas variados. Si el examen contiene ejercicios apenas distintos, no puede resolverlos. No porque haya memorizado demasiado, sino porque no incorporó suficiente estructura conceptual.

Un modelo subajustado se parece a ese estudiante: no aprendió los patrones principales del problema.

También puede ocurrir que el modelo tenga suficiente capacidad, pero los datos estén mal preparados. Si una variable categórica importante no fue codificada correctamente, si los valores faltantes fueron tratados de forma inadecuada o si una variable numérica quedó en un formato incorrecto, el modelo puede no aprovechar la información disponible.

Por ejemplo, si una columna de fechas se usa como texto sin extraer información útil, quizá el modelo no capture patrones de mes, día de la semana o temporada. Si una columna de barrio se elimina por completo, tal vez se pierda una señal importante para predecir precios.

```text id="uujlz5"
A veces el problema no es el algoritmo,
sino la representación de los datos.
```

Para reducir el subajuste, podemos probar distintas estrategias. Una es agregar variables relevantes. Si el modelo no tiene suficiente información, mejorar las variables de entrada puede ayudar. Otra es usar un modelo más flexible, capaz de capturar relaciones no lineales o interacciones. También podemos revisar el preprocesamiento, corregir errores, construir nuevas características o reformular el problema.

```text id="tgwn9j"
Para reducir subajuste podemos:
agregar variables relevantes,
mejorar el preprocesamiento,
crear mejores características,
usar un modelo más flexible,
revisar la formulación del problema.
```

Pero debemos hacerlo con cuidado. Si aumentamos demasiado la complejidad, podemos pasar del subajuste al sobreajuste. Mejorar un modelo no significa hacerlo cada vez más complejo sin límite. Significa darle la capacidad suficiente para aprender patrones reales, pero seguir evaluando si generaliza.

Por ejemplo, si una regresión lineal simple funciona mal, podemos probar una regresión lineal múltiple con más variables. Si todavía falla porque la relación es no lineal, más adelante podríamos considerar árboles, Random Forest, Gradient Boosting u otros modelos más flexibles. Pero cada paso debe evaluarse con datos de prueba o validación.

El subajuste también nos recuerda que no todos los problemas pueden resolverse solo cambiando el modelo. Si los datos no contienen información suficiente, ningún algoritmo podrá producir buenas predicciones. Si la variable objetivo está mal definida, si los registros son inconsistentes o si el fenómeno es muy aleatorio, el margen de mejora puede ser limitado.

```text id="mli5ev"
Un modelo no puede aprender
una señal que no está en los datos.
```

Por eso, ante un posible subajuste, no conviene culpar automáticamente al algoritmo. Debemos revisar el problema completo: la pregunta, la variable objetivo, las variables predictoras, la calidad de los datos, el preprocesamiento y las métricas usadas para evaluar.

También es importante comparar con una referencia simple. Si nuestro modelo funciona apenas igual que predecir siempre la clase mayoritaria o el promedio de la variable objetivo, probablemente no esté aprendiendo mucho. Una mejora clara sobre esa referencia es una señal mínima de que el modelo está capturando alguna relación útil.

```text id="eal4vd"
Si un modelo no mejora una referencia simple,
debemos revisar si realmente está aprendiendo.
```

> El underfitting o subajuste ocurre cuando un modelo no logra aprender suficientemente los patrones del problema. Suele manifestarse como mal desempeño tanto en entrenamiento como en prueba. Puede deberse a modelos demasiado simples, variables insuficientes, mala preparación de datos, relaciones no lineales no capturadas o problemas en la formulación. Reducir el subajuste implica mejorar la información disponible o aumentar la capacidad del modelo, siempre cuidando no caer en sobreajuste.


## Modelos demasiado simples

Un modelo demasiado simple es aquel que no tiene suficiente capacidad para representar los patrones importantes del problema. Puede aprender una tendencia muy general, pero no capturar detalles relevantes de la relación entre las variables de entrada y la variable objetivo.

Este tipo de modelo suele estar asociado al subajuste. No falla porque haya aprendido demasiado los datos de entrenamiento, sino porque no aprende lo suficiente. Sus predicciones son pobres incluso sobre los datos que usó para entrenar.

```text id="p65okq"
Modelo demasiado simple:
no logra capturar los patrones principales
del problema.
```

Un ejemplo claro aparece en regresión. Supongamos que queremos predecir el consumo eléctrico de un edificio usando solo la temperatura exterior. Además, imaginemos que el consumo aumenta en días muy fríos por calefacción y también en días muy calurosos por refrigeración. En temperaturas templadas, el consumo es menor.

La relación real podría tener forma de curva: alta en los extremos y baja en el centro. Si intentamos representarla con una regresión lineal simple, el modelo usará una recta. Esa recta difícilmente capture bien el patrón.

```text id="xxuytn"
Relación real:
el consumo aumenta con frío intenso
y también con calor intenso.

Modelo demasiado simple:
una sola recta.
```

En este caso, el modelo puede equivocarse de manera sistemática. Tal vez subestime el consumo en días extremos y lo sobreestime en días templados. No importa cuánto lo entrenemos: si la forma del modelo no permite representar esa curva, el problema persistirá.

Algo similar puede ocurrir en clasificación. Imaginemos que queremos separar dos clases que se mezclan de manera no lineal. Si usamos un modelo que solo puede trazar una frontera de decisión muy simple, quizá no logre separar adecuadamente los grupos.

Por ejemplo, una regresión logística básica puede funcionar muy bien cuando las clases se separan aproximadamente mediante una frontera lineal. Pero si la separación real tiene una forma más compleja, el modelo puede quedarse corto.

```text id="vxwnzk"
Frontera simple:
puede funcionar si las clases se separan claramente.

Frontera compleja:
puede requerir un modelo más flexible.
```

Esto no significa que los modelos simples sean malos. Al contrario, muchas veces son una excelente primera opción. Son más fáciles de entrenar, interpretar, comunicar y controlar. Una regresión lineal o una regresión logística pueden ser muy útiles si la relación principal es razonablemente simple.

El problema aparece cuando la simplicidad impide aprender lo necesario.

```text id="xwxka8"
Modelo simple:
puede ser útil y preferible.

Modelo demasiado simple:
no alcanza para representar el problema.
```

La diferencia no se decide por el nombre del algoritmo, sino por su desempeño y por el análisis de errores. Un modelo simple puede ser suficiente en un problema y totalmente insuficiente en otro. La pregunta no es si el modelo es simple, sino si su simplicidad alcanza para capturar los patrones relevantes.

Una señal de modelo demasiado simple es el mal desempeño tanto en entrenamiento como en prueba. Si el modelo no logra buenos resultados ni siquiera con los datos que vio, probablemente no tiene suficiente capacidad, no cuenta con variables adecuadas o la relación entre variables y objetivo no está bien representada.

```text id="kg6zsk"
Entrenamiento: mal desempeño.
Prueba: mal desempeño parecido.

Posible diagnóstico:
modelo demasiado simple o datos poco informativos.
```

Por ejemplo, en regresión:

```text id="z2j6fi"
MAE en entrenamiento:  35
MAE en prueba:         37
```

Si esos errores son altos para el problema, el modelo no está aprendiendo lo suficiente. La diferencia entre entrenamiento y prueba no es el problema principal. El problema es que ambos resultados son malos.

En clasificación:

```text id="pa4yhq"
F1-score en entrenamiento: 0.55
F1-score en prueba:        0.53
```

Si una referencia simple obtiene resultados similares, el modelo no está aportando demasiado.

Un modelo puede ser demasiado simple por varias razones. Una es que use pocas variables. Si queremos predecir precios de viviendas usando solo superficie, podemos capturar una parte del problema, pero no toda. Si la ubicación es fundamental y no está incluida, el modelo no podrá usar esa información.

```text id="zupijw"
Faltan variables relevantes:
el modelo puede quedar limitado,
aunque el algoritmo esté bien usado.
```

Otra razón es que las variables estén representadas de manera pobre. Tal vez tenemos una fecha, pero no extraemos el mes, el día de la semana o si se trata de un feriado. Tal vez tenemos una dirección, pero no transformamos esa información en zona, distancia o características del barrio. Tal vez tenemos texto, pero lo ignoramos por completo.

En esos casos, el modelo parece simple no solo por el algoritmo, sino por la forma en que le presentamos los datos.

```text id="jfn98h"
Un modelo aprende de las variables que recibe.
Si la representación es pobre,
el aprendizaje también puede ser pobre.
```

También puede ser demasiado simple porque sus restricciones son fuertes. Un árbol de decisión con profundidad máxima de 1 solo puede hacer una pregunta antes de decidir. A ese tipo de árbol se lo suele llamar “stump” o tocón de decisión. Puede ser útil como componente de algunos métodos, pero por sí solo puede resultar insuficiente para problemas complejos.

```text id="x2rx57"
Árbol con profundidad 1:
solo puede dividir los datos una vez.

Si el problema requiere varias decisiones,
puede subajustar.
```

En modelos lineales, la simplicidad aparece en la forma de la relación. Una regresión lineal básica suma aportes proporcionales de las variables. Si el problema necesita curvas, interacciones o cambios de comportamiento según rangos, esa forma puede quedarse corta.

Por ejemplo, el efecto de una promoción sobre las ventas puede depender del tipo de producto. La influencia de la temperatura sobre el consumo puede depender del horario. El impacto de la superficie sobre el precio puede depender del barrio. Una suma lineal básica no representa automáticamente esas interacciones.

```text id="aao30o"
Si el efecto de una variable depende de otra,
un modelo demasiado simple puede no capturarlo.
```

Para mejorar un modelo demasiado simple, una posibilidad es agregar variables relevantes. Si el modelo no tiene suficiente información, darle mejores datos puede ayudar más que cambiar de algoritmo.

Otra posibilidad es construir mejores características. Por ejemplo, transformar fechas en variables útiles, agrupar categorías, crear indicadores, calcular promedios históricos o representar relaciones que el modelo no capturaría por sí solo.

También podemos usar un modelo más flexible. Si una regresión lineal no alcanza, más adelante podríamos probar árboles, Random Forest, Gradient Boosting u otros métodos. Si una frontera lineal no separa bien las clases, podríamos probar modelos capaces de representar fronteras más complejas.

```text id="kmpkp5"
Para enfrentar un modelo demasiado simple:
agregar variables útiles,
mejorar la representación,
usar modelos más flexibles,
revisar la formulación del problema.
```

Pero este aumento de complejidad debe hacerse con cuidado. Un modelo más flexible puede reducir el subajuste, pero también aumentar el riesgo de sobreajuste. La mejora debe medirse en datos no vistos, no solo en entrenamiento.

Por ejemplo, si un árbol de decisión poco profundo funciona mal, podemos permitir mayor profundidad. Al principio, eso puede mejorar tanto entrenamiento como prueba. Pero si seguimos aumentando la profundidad sin control, el árbol puede empezar a memorizar los datos de entrenamiento y perder generalización.

```text id="mmim9z"
Aumentar complejidad puede ayudar,
pero solo hasta cierto punto.

Después, puede aparecer sobreajuste.
```

Por eso, no buscamos el modelo más simple ni el más complejo. Buscamos el nivel de complejidad adecuado para el problema y los datos disponibles.

Un modelo simple tiene ventajas claras: es más fácil de interpretar, más rápido, más estable y más fácil de comunicar. Si su desempeño es suficiente, puede ser preferible a un modelo complejo. Pero si sus errores son altos y sistemáticos, necesitamos darle más capacidad o mejor información.

La decisión debe basarse en evidencia. Primero evaluamos el modelo. Luego analizamos sus errores. Si vemos que falla tanto en entrenamiento como en prueba, y que no mejora una referencia simple, sospechamos subajuste. A partir de ahí, revisamos si el modelo es demasiado simple, si faltan variables o si el problema está mal planteado.

> Un modelo demasiado simple es aquel que no alcanza a capturar los patrones relevantes de los datos. Suele producir mal desempeño tanto en entrenamiento como en prueba y se relaciona con el subajuste. La solución puede implicar agregar variables, mejorar la representación de los datos o usar modelos más flexibles. Sin embargo, aumentar la complejidad siempre debe evaluarse con cuidado para no pasar al extremo opuesto: el sobreajuste.


## Modelos demasiado complejos

Un modelo demasiado complejo es aquel que tiene tanta flexibilidad que puede adaptarse en exceso a los datos de entrenamiento. En lugar de aprender solo los patrones generales, también puede capturar ruido, irregularidades, valores extremos o coincidencias accidentales. Cuando eso ocurre, el modelo puede funcionar muy bien con los datos conocidos, pero peor con datos nuevos.

Este tipo de modelo se relaciona directamente con el sobreajuste.

```text id="osn6ld"
Modelo demasiado complejo:
aprende muy bien el entrenamiento,
pero puede perder capacidad de generalización.
```

La complejidad de un modelo no siempre es fácil de ver a simple vista. No depende únicamente de cuántas líneas de código usamos o de si el algoritmo parece avanzado. Depende de cuánta libertad tiene el modelo para ajustarse a los datos.

Un árbol de decisión muy profundo, por ejemplo, puede ser muy complejo. Si permitimos que siga dividiendo los datos una y otra vez, puede terminar creando reglas muy específicas para pequeños grupos de casos. Incluso puede llegar a hojas con uno o muy pocos ejemplos. En entrenamiento, eso puede producir resultados excelentes. Pero esas reglas pueden no servir para casos nuevos.

```text id="v656hi"
Árbol muy profundo:
muchas decisiones específicas,
posible memorización de casos particulares.
```

Supongamos un modelo para predecir abandono de clientes. Un árbol demasiado profundo podría aprender reglas como:

```text id="vslvba"
si el cliente tiene 37 años,
vive en cierta zona,
hizo 2 reclamos,
compró cierto producto,
y se conectó exactamente 4 veces,
entonces abandona.
```

Tal vez esa regla funcione para algunos clientes del entrenamiento, pero no necesariamente representa un patrón estable. Puede ser una coincidencia. Si el modelo construye muchas reglas de ese tipo, corre riesgo de sobreajustar.

En regresión ocurre algo parecido. Un modelo muy flexible puede seguir demasiado de cerca los puntos del entrenamiento. Si los datos tienen ruido, la curva aprendida puede ondular para acercarse a cada observación, pero perder la tendencia general. Visualmente, sería una línea que parece encajar muy bien con los puntos conocidos, pero que se comporta de manera extraña entre ellos o frente a nuevos casos.

```text id="tmle88"
Ajustarse demasiado a los puntos observados
puede producir malas predicciones
para puntos nuevos.
```

Esto no significa que la complejidad sea mala en sí misma. Muchos problemas reales necesitan modelos flexibles. Si la relación entre variables es no lineal, si hay interacciones importantes o si la frontera de decisión es compleja, un modelo simple puede quedarse corto. En esos casos, aumentar la complejidad puede mejorar la generalización.

El problema aparece cuando la complejidad supera lo que los datos pueden sostener.

```text id="ro4ok5"
Complejidad útil:
captura patrones reales.

Complejidad excesiva:
captura ruido y detalles accidentales.
```

La cantidad de datos disponibles influye mucho. Con pocos ejemplos, un modelo complejo tiene más riesgo de encontrar patrones casuales. Si hay muchas variables y pocas filas, el riesgo aumenta aún más. Algunas combinaciones de variables pueden parecer predictivas por azar, aunque no funcionen en datos nuevos.

```text id="gjwmqe"
Pocos datos + modelo flexible
→ mayor riesgo de sobreajuste.
```

También influye la calidad de los datos. Si hay errores de carga, etiquetas incorrectas, valores extremos no revisados o ruido fuerte, un modelo complejo puede intentar aprender esas irregularidades. Un modelo más simple quizá ignore parte de ese ruido y capture una tendencia más estable.

Por eso, ante un modelo demasiado complejo, no siempre la solución es cambiar inmediatamente de algoritmo. A veces conviene revisar primero los datos, las variables y el proceso de evaluación. Tal vez el problema no sea solo la complejidad del modelo, sino la información con la que se lo alimenta.

Una señal típica de exceso de complejidad es una gran diferencia entre el desempeño en entrenamiento y el desempeño en prueba. Por ejemplo:

```text id="rnnpdb"
Entrenamiento:
MAE = 1.5

Prueba:
MAE = 22.0
```

O en clasificación:

```text id="gwp93c"
Entrenamiento:
F1-score = 0.98

Prueba:
F1-score = 0.61
```

En estos casos, el modelo parece muy bueno cuando evalúa los datos que ya vio, pero no conserva ese desempeño en datos no vistos. Esa brecha es una señal fuerte de sobreajuste.

También puede aparecer una mejora engañosa cuando miramos solo entrenamiento. Si aumentamos la complejidad de un modelo, muchas veces el desempeño en entrenamiento mejora. Un árbol más profundo, más variables o más flexibilidad pueden reducir errores sobre los datos conocidos. Pero esa mejora no siempre se traslada al conjunto de prueba.

```text id="zpr00i"
Más complejidad casi siempre puede mejorar entrenamiento.
Lo importante es si mejora datos no vistos.
```

Esta es una idea central. No elegimos el modelo que mejor ajusta el entrenamiento. Elegimos el modelo que mejor generaliza.

Un ejemplo claro aparece con árboles de decisión. Si aumentamos `max_depth`, el árbol puede hacer más preguntas y construir reglas más específicas. Al principio, eso puede ayudar: el modelo deja de ser demasiado simple y captura patrones importantes. Pero, pasado cierto punto, más profundidad puede empezar a memorizar detalles del entrenamiento.

```text id="zyfps7"
Profundidad baja:
riesgo de subajuste.

Profundidad intermedia:
posible buen equilibrio.

Profundidad muy alta:
riesgo de sobreajuste.
```

Este tipo de comportamiento se observa también con otros hiperparámetros. Muchos modelos tienen controles que regulan su flexibilidad. En KNN, un valor muy bajo de `k` puede producir decisiones muy sensibles a casos individuales. En regresión lineal con muchas variables irrelevantes, el modelo puede ajustarse a ruido. En modelos más avanzados, también habrá parámetros que controlen la complejidad.

No necesitamos estudiar todos esos casos ahora. Lo importante es entender la idea general: la complejidad permite aprender patrones más ricos, pero también aumenta el riesgo de aprender detalles que no generalizan.

Para reducir el riesgo de modelos demasiado complejos, podemos usar distintas estrategias. Una es limitar la flexibilidad del modelo. En árboles, podemos limitar la profundidad, exigir más muestras por hoja o impedir divisiones con poca información. En otros modelos, podemos usar regularización o ajustar hiperparámetros que restrinjan el ajuste.

```text id="catdla"
Controlar complejidad:
dar al modelo suficiente libertad para aprender,
pero no tanta como para memorizar.
```

Otra estrategia es mejorar la evaluación. Si usamos solo una separación train/test, podemos tener una idea inicial, pero tal vez dependa demasiado de esa partición. Más adelante estudiaremos validación cruzada, que permite evaluar de manera más estable usando varias divisiones de los datos.

También podemos reducir variables irrelevantes, revisar outliers, mejorar el preprocesamiento o conseguir más datos. Más datos de buena calidad suelen ayudar a los modelos complejos, porque les dan más ejemplos para distinguir patrones reales de coincidencias accidentales.

```text id="no5qwj"
Para controlar modelos demasiado complejos podemos:
limitar hiperparámetros,
usar regularización,
mejorar datos,
eliminar ruido,
evaluar con validación,
comparar entrenamiento y prueba.
```

De todos modos, simplificar no significa elegir siempre el modelo más básico. Si simplificamos demasiado, podemos volver al subajuste. El objetivo es encontrar un equilibrio entre capacidad y generalización.

Un modelo demasiado simple no aprende suficiente. Un modelo demasiado complejo aprende demasiado de los detalles del entrenamiento. Un buen modelo aprende lo necesario para resolver el problema en datos nuevos.

```text id="rz8sv8"
Modelo demasiado simple:
alto error en entrenamiento y prueba.

Modelo demasiado complejo:
bajo error en entrenamiento,
alto error en prueba.

Modelo adecuado:
error razonable en entrenamiento
y buen desempeño en prueba.
```

También conviene considerar la interpretabilidad. Los modelos complejos pueden ser más difíciles de explicar. Si la mejora en métricas es pequeña, tal vez no justifique perder claridad. En cambio, si la mejora es grande y el problema lo requiere, puede tener sentido aceptar mayor complejidad, siempre que la evaluación sea sólida.

Por ejemplo, una regresión lineal puede ser preferible a un modelo muy complejo si sus errores son apenas mayores, pero sus resultados son más fáciles de comunicar. En cambio, si un modelo más flexible reduce mucho los errores en datos no vistos, puede ser la mejor opción para una tarea predictiva.

La decisión no es “simple contra complejo”. La decisión es qué nivel de complejidad necesita el problema y qué costo tiene esa complejidad.

```text id="sibwfb"
La complejidad debe justificarse
por una mejora real en datos no vistos
y por el uso que tendrá el modelo.
```

> Un modelo demasiado complejo tiene tanta flexibilidad que puede adaptarse a detalles específicos del conjunto de entrenamiento y perder capacidad de generalización. Suele mostrar muy buen desempeño en entrenamiento, pero peor desempeño en prueba. La complejidad puede ser necesaria para capturar patrones reales, pero debe controlarse mediante hiperparámetros, regularización, buena evaluación y análisis de errores. El objetivo no es usar el modelo más complejo, sino el que logre el mejor equilibrio entre aprendizaje, generalización e interpretabilidad.


## Señales de alerta

El sobreajuste y el subajuste no siempre se ven de manera inmediata. Un modelo puede producir métricas, gráficos y predicciones que parecen razonables a primera vista, pero aun así tener problemas importantes. Por eso, al evaluar modelos, conviene prestar atención a ciertas señales de alerta.

La primera señal aparece al comparar el desempeño en entrenamiento y en prueba. Si el modelo funciona muy bien en entrenamiento, pero mucho peor en prueba, debemos sospechar sobreajuste.

```text id="g7ybta"
Entrenamiento: muy buen desempeño.
Prueba: desempeño mucho peor.

Posible problema:
sobreajuste.
```

Por ejemplo, en clasificación:

```text id="v0dnxn"
Accuracy en entrenamiento: 0.98
Accuracy en prueba:        0.71
```

O en regresión:

```text id="qwvwit"
MAE en entrenamiento:  3.2
MAE en prueba:        28.5
```

En ambos casos, el modelo parece haber aprendido muy bien los datos de entrenamiento, pero no logra mantener ese rendimiento en datos nuevos. La diferencia entre ambos resultados es demasiado grande como para ignorarla.

No existe un número universal que indique cuándo la diferencia es aceptable y cuándo no. Depende del problema, del tamaño del dataset, de la métrica y de la variabilidad natural de los datos. Pero la idea general es clara: una brecha grande entre entrenamiento y prueba debe investigarse.

La segunda señal aparece cuando el modelo funciona mal tanto en entrenamiento como en prueba. En ese caso, el problema probablemente no sea sobreajuste, sino subajuste o falta de señal en los datos.

```text id="yugkdb"
Entrenamiento: mal desempeño.
Prueba: mal desempeño parecido.

Posible problema:
subajuste,
variables insuficientes
o problema mal formulado.
```

Por ejemplo:

```text id="nzcltz"
MAE en entrenamiento:  42.0
MAE en prueba:         44.3
```

Si ese nivel de error es alto para el problema, el modelo no está aprendiendo lo necesario. La cercanía entre entrenamiento y prueba no significa que el modelo sea bueno. Solo indica que falla de manera parecida en ambos conjuntos.

En clasificación, una señal parecida podría ser:

```text id="ex0b2a"
F1-score en entrenamiento: 0.48
F1-score en prueba:        0.46
```

Si una referencia simple obtiene resultados similares, el modelo no está aportando mucho.

La tercera señal de alerta aparece cuando el modelo no mejora una estrategia básica. En regresión, una referencia simple puede ser predecir siempre el promedio o la mediana de la variable objetivo. En clasificación, puede ser predecir siempre la clase mayoritaria. Si el modelo entrenado apenas mejora esa referencia, debemos revisar si realmente está aprendiendo patrones útiles.

```text id="xhqfsu"
Referencia simple:
MAE = 20.0

Modelo entrenado:
MAE = 19.4
```

La mejora existe, pero es pequeña. Tal vez no justifique la complejidad del modelo. También puede indicar que las variables predictoras tienen poca señal, que el problema está mal planteado o que el preprocesamiento no está aprovechando la información disponible.

La cuarta señal aparece cuando los resultados son demasiado buenos para ser ciertos. En problemas reales, un desempeño casi perfecto merece una revisión cuidadosa. Puede ser real en problemas muy simples o datasets muy limpios, pero muchas veces indica una evaluación incorrecta.

```text id="qdbj3s"
Accuracy = 1.00
F1-score = 1.00
MAE ≈ 0
R² ≈ 1.00

Señal:
revisar antes de confiar.
```

Un resultado perfecto puede deberse a fuga de datos, evaluación sobre el conjunto de entrenamiento, duplicados entre entrenamiento y prueba, variables que revelan la respuesta o errores en la separación de datos. No debemos celebrar automáticamente un resultado perfecto sin revisar cómo se obtuvo.

La quinta señal son métricas muy diferentes entre clases o grupos. Un modelo puede tener una métrica global aceptable y, sin embargo, fallar en una clase minoritaria o en un segmento importante.

Por ejemplo, en clasificación:

```text id="um2zhf"
Clase mayoritaria:
recall = 0.95

Clase minoritaria:
recall = 0.30
```

La accuracy global podría parecer buena, pero el modelo está fallando en la clase que quizá más importa. Esto es frecuente en problemas con desbalance de clases.

En regresión también puede ocurrir algo parecido. El modelo puede tener buen error promedio, pero fallar mucho en valores altos, en ciertas zonas, en determinados productos o en períodos específicos.

```text id="snb08x"
MAE general: 8 unidades

MAE en días normales: 5 unidades
MAE en días de promoción: 35 unidades
```

El promedio general oculta que el modelo falla justamente en un segmento relevante. Por eso, las métricas globales deben complementarse con análisis por grupos, rangos o condiciones importantes.

La sexta señal aparece en los residuos o errores. Si los errores muestran un patrón claro, el modelo puede estar dejando información sin capturar. En regresión, por ejemplo, si el modelo subestima sistemáticamente los valores altos y sobreestima los valores bajos, puede estar usando una forma demasiado simple.

```text id="k5js72"
Errores con patrón:
posible relación no capturada,
variables faltantes
o modelo inadecuado.
```

Un buen análisis de errores no busca solo calcular una métrica. Busca entender dónde y cómo se equivoca el modelo. Si los errores se concentran en una zona, un grupo o un tipo de caso, hay una pista para mejorar.

La séptima señal de alerta es la inestabilidad. Si pequeños cambios en los datos producen resultados muy diferentes, el modelo puede ser demasiado sensible. Esto puede ocurrir con datasets pequeños, modelos muy flexibles o datos con mucho ruido.

Por ejemplo, si al cambiar la separación train/test el modelo pasa de un R² de 0.82 a uno de 0.35, conviene investigar. Tal vez el conjunto de datos sea pequeño, la partición sea poco representativa o el modelo dependa demasiado de algunos casos particulares.

```text id="r58tuc"
Resultados muy variables
ante pequeñas modificaciones
→ posible falta de estabilidad.
```

Más adelante, la validación cruzada nos ayudará a evaluar esta variabilidad con más criterio. Por ahora, alcanza con reconocer que una única métrica obtenida en una única separación no siempre cuenta toda la historia.

La octava señal aparece cuando el modelo depende demasiado de variables sospechosas. Si una variable tiene una importancia enorme o un coeficiente inesperadamente dominante, debemos revisar si esa variable es legítima. Tal vez contiene información futura, resume indirectamente la respuesta o no estaría disponible al momento de predicción.

Por ejemplo, en un modelo de abandono de clientes, una variable como `fecha_baja` no debería usarse para predecir si el cliente abandonará. En un modelo de aprobación de créditos, una variable calculada después de la decisión tampoco sería válida. Si el modelo se apoya en ese tipo de columnas, puede parecer excelente en evaluación, pero fallar en uso real.

```text id="wti5wp"
Variable sospechosa:
muy predictiva,
pero no disponible antes de la predicción
o demasiado cercana a la respuesta.
```

Esta señal conecta directamente con la fuga de datos, que retomaremos al final del capítulo.

La novena señal aparece cuando el desempeño en entrenamiento mejora al aumentar complejidad, pero el desempeño en prueba empeora. Este comportamiento es típico cuando cruzamos el punto de equilibrio y empezamos a sobreajustar.

Por ejemplo, en un árbol de decisión:

```text id="nq0nq5"
Profundidad 3:
prueba razonable.

Profundidad 8:
mejor entrenamiento,
prueba similar.

Profundidad 20:
entrenamiento casi perfecto,
prueba peor.
```

El modelo se vuelve cada vez más capaz de memorizar entrenamiento, pero eso no mejora la generalización. En ese caso, conviene limitar la complejidad.

La décima señal es la falta de coherencia con el conocimiento del problema. Si el modelo produce predicciones absurdas, coeficientes difíciles de justificar o reglas que contradicen de manera extraña lo esperable, no necesariamente está mal, pero merece revisión.

Por ejemplo, un modelo de consumo que predice valores negativos, un modelo de precios que asigna enorme importancia a un identificador interno o un modelo de ventas que ignora por completo una variable claramente relevante pueden indicar problemas de datos, preprocesamiento o formulación.

```text id="q9g53b"
Un modelo puede tener métricas aceptables
y aun así producir señales incoherentes.
```

La evaluación técnica debe dialogar con el sentido del problema. Las métricas son necesarias, pero no sustituyen la revisión conceptual.

Estas señales de alerta no deben interpretarse como diagnósticos automáticos. Una brecha entre entrenamiento y prueba puede indicar sobreajuste, pero debemos confirmar. Un mal desempeño general puede indicar subajuste, pero también datos insuficientes o un problema muy difícil. Un resultado perfecto puede tener explicación legítima, aunque siempre conviene revisar.

Lo importante es usar estas señales como disparadores de análisis.

```text id="x1ym45"
Una señal de alerta no cierra el diagnóstico.
Abre una pregunta.
```

> Al evaluar modelos debemos mirar más que una métrica final. Señales como gran diferencia entre entrenamiento y prueba, mal desempeño en ambos conjuntos, resultados demasiado perfectos, errores concentrados en ciertos grupos, patrones en residuos, inestabilidad o dependencia de variables sospechosas pueden indicar sobreajuste, subajuste, fuga de datos o problemas de formulación. Detectar estas señales a tiempo permite corregir el modelo antes de confiar en predicciones que no generalizan.


## Resultados demasiado buenos para ser ciertos

En Machine Learning, obtener buenos resultados es el objetivo. Queremos modelos que clasifiquen correctamente, que predigan valores con poco error y que generalicen bien a datos nuevos. Sin embargo, cuando los resultados son **demasiado buenos**, especialmente en problemas reales, conviene detenerse y revisar.

Un desempeño perfecto o casi perfecto no siempre significa que encontramos un gran modelo. A veces significa que hay un problema en la forma en que preparamos los datos, entrenamos, evaluamos o definimos las variables.

```text id="ryeafl"
Resultado muy alto:
puede ser una buena señal.

Resultado perfecto o sospechosamente alto:
debe revisarse con cuidado.
```

Por ejemplo, en clasificación podríamos obtener:

```text id="fbx4i7"
accuracy = 1.00
precision = 1.00
recall = 1.00
F1-score = 1.00
```

O en regresión:

```text id="gv5kgv"
MAE = 0
RMSE = 0
R² = 1.00
```

En un problema real con datos ruidosos, variables incompletas y situaciones diversas, resultados así son poco frecuentes. No son imposibles, pero sí merecen una revisión cuidadosa antes de confiar en ellos.

Una primera explicación posible es que estemos evaluando el modelo sobre los mismos datos que usamos para entrenarlo. Si entrenamos y evaluamos con el mismo conjunto, el resultado puede ser demasiado optimista. El modelo no está demostrando que generaliza; solo está mostrando cómo se comporta sobre datos conocidos.

```text id="a9cupz"
Entrenar y evaluar con los mismos datos
puede producir una evaluación engañosa.
```

Esto es especialmente grave con modelos flexibles. Un árbol de decisión muy profundo, por ejemplo, puede memorizar los datos de entrenamiento y obtener un desempeño casi perfecto sobre ellos. Pero eso no significa que funcionará bien en datos nuevos.

Por eso, siempre debemos preguntarnos:

```text id="l5znd0"
¿La métrica fue calculada sobre datos de prueba?
¿El modelo ya había visto esos casos durante el entrenamiento?
```

Una segunda explicación posible es la fuga de datos. La fuga de datos ocurre cuando el modelo recibe, directa o indirectamente, información que no estaría disponible en el momento real de predicción. Esa información puede revelar la respuesta y hacer que el modelo parezca mucho mejor de lo que realmente sería.

Por ejemplo, si queremos predecir abandono de clientes, no deberíamos incluir una columna como `fecha_baja` o `motivo_baja`. Esas variables se conocen después de que el cliente abandona. Si aparecen en el entrenamiento, el modelo tendrá una pista directa sobre la respuesta.

```text id="n3ggap"
Problema:
predecir si un cliente abandonará.

Variable sospechosa:
fecha_baja.

Motivo:
solo se conoce después del abandono.
```

En un modelo de crédito, tampoco deberíamos usar variables generadas después de aprobar o rechazar la solicitud si el objetivo es predecir el riesgo antes de tomar la decisión. En un modelo de diagnóstico, no deberíamos usar resultados de estudios posteriores si el modelo se supone que debe funcionar antes de esos estudios.

La fuga de datos puede producir resultados sorprendentemente altos porque el modelo no está aprendiendo patrones legítimos. Está usando información que, en la práctica, no debería tener.

Una tercera explicación posible son duplicados o casos casi idénticos repartidos entre entrenamiento y prueba. Si el mismo registro aparece en ambos conjuntos, el modelo puede “reconocer” ejemplos que supuestamente eran nuevos. Esto vuelve la evaluación demasiado optimista.

```text id="w0lbsg"
Si un mismo caso aparece en entrenamiento y prueba,
la prueba deja de representar datos realmente no vistos.
```

Esto puede ocurrir con registros duplicados, múltiples filas de una misma persona, mediciones repetidas o datos generados a partir del mismo evento. En esos casos, separar filas al azar puede no ser suficiente. Tal vez haya que separar por cliente, por paciente, por producto, por fecha o por alguna unidad que evite mezclar información relacionada.

Por ejemplo, si tenemos varias transacciones de un mismo cliente y algunas quedan en entrenamiento mientras otras quedan en prueba, el modelo puede aprender características específicas de ese cliente. Según el objetivo, eso puede ser aceptable o puede inflar la evaluación. Si queremos medir desempeño sobre clientes completamente nuevos, debemos separar por cliente.

Una cuarta explicación es que la variable objetivo esté codificada de forma indirecta en alguna columna. A veces no hay una columna tan evidente como `fecha_baja`, pero sí existe una variable que resume o deriva de la respuesta.

Por ejemplo, en un dataset de abandono, una columna como `cantidad_dias_desde_ultima_actividad` puede ser válida si se calcula antes del momento de predicción. Pero puede ser problemática si fue calculada usando información posterior al abandono. La misma variable puede ser legítima o ilegítima según cómo y cuándo se construyó.

```text id="o8tzw5"
Una variable no es válida solo por su nombre.
Hay que saber cómo fue calculada
y en qué momento estaría disponible.
```

Esta es una de las razones por las que el contexto del problema importa tanto. No alcanza con mirar la tabla. Debemos entender la línea temporal de los datos.

Una quinta explicación posible es que el problema sea demasiado fácil por una construcción artificial del dataset. Algunos datasets educativos están diseñados para mostrar conceptos y pueden tener patrones muy claros. En esos casos, un desempeño muy alto puede ser razonable. Pero aun así conviene distinguir entre un ejemplo didáctico y un problema real.

```text id="ysv7mk"
Dataset educativo:
puede tener patrones limpios y simples.

Dataset real:
suele tener ruido, excepciones y cambios.
```

Esto no invalida los datasets didácticos. Son muy útiles para aprender. Pero no debemos extrapolar su facilidad a problemas reales sin revisión.

Otra causa de resultados sospechosos es usar el conjunto de prueba muchas veces para tomar decisiones. Si entrenamos muchos modelos, probamos muchas configuraciones y elegimos la que mejor funciona en el test, el conjunto de prueba deja de ser una evaluación independiente. De manera indirecta, empezamos a ajustar nuestras decisiones a ese test.

```text id="dmzpxg"
Usar el test una y otra vez para elegir modelos
puede volver optimista la evaluación.
```

Más adelante, cuando estudiemos validación cruzada e hiperparámetros, veremos formas más adecuadas de separar entrenamiento, validación y prueba. Por ahora, lo importante es recordar que el test debe cuidarse como una evaluación honesta de datos no vistos.

Los resultados demasiado buenos también pueden aparecer por errores en el código. Por ejemplo, podríamos calcular una métrica comparando `y_test` con `y_test` en lugar de compararla con `y_pred`. O podríamos mezclar índices y terminar evaluando predicciones sobre valores que no corresponden. También podríamos aplicar una transformación incorrecta que incluye información del conjunto de prueba.

```text id="h1qugd"
Antes de confiar en un resultado perfecto,
conviene revisar el flujo completo:
datos, separación, entrenamiento, predicción y métricas.
```

En regresión, un R² de 1.00 o un error casi cero puede ser una señal muy fuerte de que algo debe revisarse, especialmente si la variable objetivo representa un fenómeno difícil de predecir. Precios, ventas, demanda, consumo y temperaturas suelen tener variabilidad natural. Un modelo perfecto en esos contextos puede indicar fuga de datos, duplicados o evaluación incorrecta.

En clasificación, una accuracy perfecta también debe interpretarse con cuidado. Si el problema es separar dos clases muy evidentes en un dataset simple, puede ser posible. Pero si estamos tratando con fraude, abandono, spam, riesgo o diagnóstico asistido, un resultado perfecto debería despertar sospechas.

Esto no significa que debamos desconfiar de todo buen resultado. Un modelo puede funcionar realmente bien. Pero cuanto más alto sea el desempeño, más importante es revisar que la evaluación sea válida.

```text id="tdfd96"
Buen resultado:
se celebra después de revisar.

Resultado perfecto:
se investiga antes de celebrar.
```

Una forma práctica de revisar resultados demasiado buenos es hacer preguntas concretas:

```text id="v5agwo"
¿Evaluamos sobre datos no vistos?
¿Hay variables que revelan la respuesta?
¿Se preprocesó antes de separar los datos?
¿Hay duplicados entre entrenamiento y prueba?
¿La variable objetivo aparece indirectamente en X?
¿El test fue usado muchas veces para elegir el modelo?
¿La métrica está calculada correctamente?
```

Estas preguntas no requieren desconfianza extrema, sino cuidado metodológico. Un modelo solo es útil si su evaluación representa razonablemente el uso futuro.

También conviene revisar ejemplos individuales. Si el modelo acierta todos los casos, podemos mirar qué variables está usando, qué importancia asigna a cada una y si sus decisiones tienen sentido. Si una variable domina completamente el modelo, debemos preguntarnos si es legítima. Si el modelo parece apoyarse en una columna administrativa, en una fecha posterior o en un identificador, puede haber un problema.

En muchos proyectos, detectar una fuga de datos a tiempo es más valioso que obtener una métrica alta. Una métrica inflada puede llevar a confiar en un modelo que fallará al aplicarse. En cambio, una evaluación más honesta permite mejorar el sistema de manera real.

```text id="j2q1fn"
Una métrica honesta, aunque sea más baja,
es más útil que una métrica perfecta pero engañosa.
```

> Los resultados demasiado buenos para ser ciertos deben revisarse con atención. Pueden deberse a un modelo realmente adecuado, pero también a evaluación sobre entrenamiento, fuga de datos, duplicados, variables que revelan la respuesta, uso repetido del conjunto de prueba o errores de código. En Machine Learning, no basta con obtener una métrica alta: necesitamos asegurarnos de que esa métrica representa capacidad de generalización y no una ventaja artificial creada por el flujo de trabajo.


## Relación con fuga de datos

La fuga de datos es una de las causas más peligrosas de evaluaciones engañosas. Puede hacer que un modelo parezca excelente durante la evaluación, pero que falle cuando se lo intenta usar en una situación real. Por eso, al estudiar sobreajuste, subajuste y generalización, debemos volver a esta idea.

La fuga de datos ocurre cuando el modelo recibe información que no debería estar disponible en el momento de hacer la predicción. Esa información puede revelar directa o indirectamente la respuesta. Como resultado, el modelo obtiene métricas muy altas, pero no porque haya aprendido patrones útiles, sino porque tuvo acceso a pistas indebidas.

```text id="l6kx1z"
Fuga de datos:
el modelo aprende usando información
que no tendría disponible en el uso real.
```

Esto se relaciona con la generalización porque una evaluación con fuga no mide realmente la capacidad del modelo para funcionar con datos nuevos. Mide su capacidad para aprovechar información que no debería estar ahí.

Supongamos que queremos predecir si un cliente abandonará un servicio. La variable objetivo es:

```text id="hagbsz"
abandona = sí / no
```

Si entre las variables de entrada incluimos una columna como `fecha_baja`, el modelo tendrá una pista directa. Los clientes que tienen una fecha de baja probablemente abandonaron. El modelo podrá lograr un desempeño altísimo, pero esa predicción no será válida, porque en el momento en que queremos anticipar el abandono esa fecha todavía no existiría.

```text id="q9u6mc"
Variable problemática:
fecha_baja

Motivo:
se conoce después de que el cliente abandona.
```

En este caso, el modelo no está prediciendo el futuro. Está leyendo una consecuencia del evento que supuestamente debía anticipar.

Algo similar puede ocurrir en modelos de crédito. Si queremos predecir el riesgo de una solicitud antes de aprobarla, no podemos usar variables que solo se conocen después de la aprobación, como comportamiento de pago posterior, mora futura o decisiones administrativas tomadas después del análisis.

```text id="n9m56x"
Momento de predicción:
antes de aprobar el crédito.

Variables inválidas:
datos generados después de la aprobación.
```

En diagnóstico asistido, si un modelo se supone que debe ayudar con una decisión inicial, no debería usar resultados de estudios que solo se realizan después de esa decisión. En ventas, si queremos predecir demanda futura, no podemos usar columnas calculadas con ventas reales futuras. En mantenimiento predictivo, si queremos anticipar una falla, no debemos incluir variables registradas después de que la falla ocurrió.

La pregunta clave siempre es temporal:

```text id="d8oipw"
¿Esta información estaría disponible
en el momento real de hacer la predicción?
```

Si la respuesta es no, la variable no debería estar en `X`.

La fuga de datos puede ser muy evidente, como una columna que contiene directamente la respuesta. Pero muchas veces es más sutil. Puede aparecer en variables derivadas, acumulados, promedios, indicadores, fechas, estados administrativos o transformaciones realizadas antes de separar entrenamiento y prueba.

Por ejemplo, una columna llamada `cantidad_reclamos_ultimos_30_dias` podría ser válida o inválida según cómo se haya calculado. Si se calcula usando solo información anterior al momento de predicción, puede ser una variable legítima. Pero si se calcula mirando un período que incluye días posteriores al evento que queremos predecir, puede introducir fuga.

```text id="g7i2ev"
Una variable puede parecer válida por su nombre,
pero ser inválida por la forma en que fue calculada.
```

Este punto es importante porque no alcanza con revisar la tabla final. Debemos entender el proceso que generó cada columna. En proyectos reales, muchas variables ya vienen construidas por sistemas anteriores, reportes o consultas. Si no sabemos qué período usan o qué información resumen, podemos incluir fuga sin darnos cuenta.

La fuga también puede aparecer durante el preprocesamiento. Por ejemplo, si imputamos valores faltantes usando todo el dataset antes de separar entrenamiento y prueba, el conjunto de entrenamiento recibe información estadística del conjunto de prueba. Lo mismo ocurre si escalamos variables, seleccionamos características o calculamos transformaciones usando todos los datos antes del split.

```text id="ttzmjk"
Preprocesar antes de separar
puede hacer que información del test
entre en el entrenamiento.
```

Por eso insistimos en que las transformaciones que aprenden algo de los datos deben ajustarse solo con el conjunto de entrenamiento. Luego se aplican al conjunto de prueba. Esta es una de las razones para usar `Pipeline`: ayuda a ordenar el flujo y reducir errores metodológicos.

La fuga de datos puede confundirse con sobreajuste porque ambos producen problemas de generalización. Pero no son exactamente lo mismo.

En el sobreajuste, el modelo aprende demasiado los detalles del conjunto de entrenamiento. El problema está en que se adapta a ruido o particularidades que no se repiten.

En la fuga de datos, el problema es que el modelo recibe información que no debería tener. Puede funcionar bien incluso en el conjunto de prueba, si esa información filtrada también está presente allí. Por eso la fuga puede ser más peligrosa: la evaluación puede parecer excelente y no mostrar la caída hasta el uso real.

```text id="q0r0mg"
Sobreajuste:
el modelo aprende demasiado los datos conocidos.

Fuga de datos:
el modelo aprende usando información indebida.
```

Un modelo con fuga puede obtener buen desempeño tanto en entrenamiento como en prueba. Eso lo hace difícil de detectar si solo miramos métricas. Por eso, ante resultados muy buenos, debemos revisar las variables y el flujo de datos, no solo las métricas.

```text id="w8lg78"
Métricas altas no garantizan
que la evaluación sea válida.
```

Por ejemplo, imaginemos un modelo de abandono con estos resultados:

```text id="gxjcph"
Accuracy en entrenamiento: 0.99
Accuracy en prueba:        0.98
```

A simple vista parece excelente. La diferencia entre entrenamiento y prueba es pequeña, así que no parece un caso clásico de sobreajuste. Pero si el modelo usa una columna que revela el abandono, la evaluación está contaminada. El problema no se ve comparando solo métricas.

Esto muestra que la fuga de datos puede producir una falsa generalización. El modelo parece generalizar porque funciona bien en prueba, pero en realidad el conjunto de prueba no representa correctamente el escenario futuro. Contiene información que tampoco estará disponible cuando se use el modelo.

```text id="a76nne"
Falsa generalización:
el modelo parece funcionar en test,
pero porque el test contiene pistas indebidas.
```

Para prevenir fuga de datos, debemos definir claramente el momento de predicción. Antes de construir variables, tenemos que saber en qué instante se supone que el modelo hará su predicción y qué información existe en ese momento.

Por ejemplo:

```text id="fc5mo3"
Predicción de abandono:
al cierre de cada mes,
con información disponible hasta ese día.

Predicción de demanda:
antes de iniciar la semana,
con datos históricos anteriores.

Predicción de riesgo crediticio:
antes de aprobar o rechazar la solicitud.
```

Una vez definido ese momento, podemos revisar cada variable:

```text id="c6kzsf"
¿Existe antes de la predicción?
¿Fue calculada solo con datos anteriores?
¿Depende directa o indirectamente de la respuesta?
¿Estaría disponible para casos nuevos?
```

Si una variable no pasa estas preguntas, debemos excluirla o reconstruirla correctamente.

Otra forma de prevenir fuga es separar los datos antes de realizar transformaciones que aprendan información estadística. Imputadores, escaladores, codificadores y selectores de variables deben ajustarse con entrenamiento. El conjunto de prueba debe permanecer como datos no vistos hasta el momento de evaluación.

```text id="oivvq0"
Ajustar transformaciones:
solo con entrenamiento.

Aplicar transformaciones:
a entrenamiento y prueba,
sin aprender del test.
```

También conviene revisar duplicados y relaciones entre registros. Si el mismo caso, cliente, paciente, producto o evento aparece en entrenamiento y prueba de formas muy parecidas, la evaluación puede estar inflada. En algunos problemas, no basta con separar filas al azar. Puede ser necesario separar por grupo o por tiempo.

Por ejemplo, si queremos evaluar desempeño sobre clientes nuevos, todas las filas de un mismo cliente deberían quedar en el mismo conjunto. Si algunas están en entrenamiento y otras en prueba, el modelo puede aprovechar información del cliente que no tendría para un cliente completamente nuevo.

```text id="z32vu9"
Cuando hay múltiples filas por entidad,
separar al azar puede no ser suficiente.
```

La fuga de datos también se relaciona con resultados demasiado buenos para ser ciertos. Un modelo con métricas perfectas o casi perfectas debe revisarse cuidadosamente. No porque sea imposible obtener buenos resultados, sino porque una de las explicaciones más frecuentes de resultados sospechosamente altos es la presencia de fuga.

En esos casos, conviene inspeccionar:

```text id="b4vmwb"
variables demasiado predictivas,
columnas calculadas después del evento,
duplicados entre entrenamiento y prueba,
preprocesamiento antes del split,
uso repetido del conjunto de prueba,
errores en el cálculo de métricas.
```

Detectar fuga puede ser incómodo porque suele reducir las métricas. Al quitar una variable filtrada o corregir el flujo, el desempeño puede bajar mucho. Pero esa caída no significa que empeoramos el modelo. Significa que dejamos de medir una ventaja artificial y empezamos a evaluar de manera más honesta.

```text id="cgqakz"
Una métrica más baja pero honesta
es mejor que una métrica alta producida por fuga.
```

Esta idea es fundamental para construir modelos confiables. En Machine Learning, el objetivo no es ganar una métrica en un entorno artificial, sino producir predicciones útiles en el contexto real de uso. Si la evaluación no respeta ese contexto, las conclusiones serán engañosas.

> La fuga de datos ocurre cuando el modelo usa información que no estaría disponible al momento real de predicción. Puede generar resultados extremadamente altos y una falsa sensación de generalización. Se relaciona con el sobreajuste porque ambos afectan la capacidad de funcionar en datos nuevos, pero la fuga puede ser más difícil de detectar solo con métricas. Para prevenirla, debemos definir el momento de predicción, revisar la disponibilidad de variables, separar correctamente los datos y aplicar transformaciones sin aprender del conjunto de prueba.


## Aprender lo suficiente, pero no demasiado

En este capítulo estudiamos una de las ideas centrales para construir modelos útiles: la generalización. Un modelo no se entrena para repetir los datos que ya conoce, sino para aplicar lo aprendido a casos nuevos. Por eso, evaluar solo el desempeño en entrenamiento no alcanza. Necesitamos saber si el modelo mantiene un comportamiento razonable cuando se enfrenta a datos que no vio.

El sobreajuste aparece cuando el modelo aprende demasiado los detalles del conjunto de entrenamiento. Puede capturar ruido, valores extremos, coincidencias accidentales o reglas muy específicas que no se repiten en datos nuevos. La señal típica es un desempeño muy bueno en entrenamiento y mucho peor en prueba.

```text id="y6zt4g"
Sobreajuste:
muy buen desempeño en entrenamiento,
peor desempeño en prueba.
```

El subajuste aparece cuando el modelo no aprende lo suficiente. Puede ser demasiado simple, recibir pocas variables relevantes, estar mal preparado o no tener capacidad para representar la relación entre las variables y la variable objetivo. La señal típica es un mal desempeño tanto en entrenamiento como en prueba.

```text id="qclvab"
Subajuste:
mal desempeño en entrenamiento,
mal desempeño en prueba.
```

Estos dos extremos ayudan a entender el equilibrio que buscamos. Un modelo demasiado simple puede ignorar patrones importantes. Un modelo demasiado complejo puede memorizar detalles irrelevantes. Un buen modelo debe tener la capacidad suficiente para aprender relaciones reales, pero no tanta flexibilidad como para adaptarse al ruido.

```text id="mkdgfp"
Modelo demasiado simple:
no aprende lo necesario.

Modelo demasiado complejo:
aprende detalles que no generalizan.

Modelo adecuado:
aprende patrones útiles para datos nuevos.
```

También vimos que las señales de alerta no deben interpretarse de manera automática, pero sí deben motivar revisión. Una gran diferencia entre entrenamiento y prueba puede indicar sobreajuste. Malos resultados en ambos conjuntos pueden indicar subajuste. Métricas casi perfectas pueden revelar fuga de datos, duplicados, evaluación incorrecta o variables que contienen la respuesta.

Los resultados demasiado buenos para ser ciertos merecen especial atención. En problemas reales, un modelo perfecto no debe celebrarse sin revisar primero el flujo completo: separación de datos, variables usadas, preprocesamiento, duplicados, momento de predicción y cálculo de métricas.

La fuga de datos es una causa especialmente peligrosa de evaluaciones engañosas. Ocurre cuando el modelo recibe información que no estaría disponible en el momento real de hacer la predicción. Esto puede producir métricas muy altas, incluso en prueba, pero no representa verdadera capacidad de generalización.

```text id="liqdyx"
Una métrica alta no garantiza
que la evaluación sea válida.

Primero hay que verificar
que el modelo no haya recibido información indebida.
```

Para prevenir estos problemas, es fundamental definir con claridad el momento de predicción, separar correctamente los datos, ajustar transformaciones solo con entrenamiento, cuidar el conjunto de prueba y revisar si las variables estarían realmente disponibles en el uso futuro del modelo.

> Construir modelos útiles implica aprender lo suficiente, pero no demasiado. El objetivo no es obtener el mejor resultado posible sobre los datos conocidos, sino lograr un desempeño confiable en datos nuevos. Sobreajuste, subajuste y fuga de datos son riesgos distintos, pero todos afectan la confianza que podemos tener en un modelo. Reconocerlos nos prepara para el próximo paso: evaluar de manera más estable usando validación cruzada.
