# Capítulo 28 · Interpretar modelos

## ¿Qué significa interpretar un modelo?

Entrenar un modelo y obtener una métrica aceptable no siempre alcanza. En muchos problemas también necesitamos entender, al menos en parte, cómo llega el modelo a sus predicciones, qué variables parecen influir más, qué patrones está usando y qué límites tienen esas interpretaciones. A eso nos referimos cuando hablamos de interpretar modelos.

Interpretar un modelo significa intentar explicar su comportamiento. No necesariamente significa conocer cada detalle interno del algoritmo, ni convertirlo en una regla perfecta y sencilla. Significa construir una comprensión razonable de cómo el modelo relaciona las variables de entrada con las predicciones que produce.

Interpretar un modelo: entender qué señales usa, cómo influyen las variables y qué tan confiables son esas explicaciones.

Por ejemplo, si entrenamos un modelo para predecir abandono de clientes, no solo podría interesarnos saber si obtiene buen recall o buen F1-score. También podríamos querer saber qué factores están asociados con mayor riesgo de abandono: poca actividad reciente, baja frecuencia de compra, muchos reclamos, menor antigüedad o cambios en el uso del servicio.

| Concepto | Descripción |
|---|---|
| Predicción | este cliente puede abandonar. |
| Interpretación | ¿qué variables parecen influir en esa predicción? |

Esta diferencia es importante. Una predicción responde qué espera el modelo para un caso. Una interpretación intenta explicar por qué el modelo produjo esa predicción o qué patrones generales aprendió.

En algunos contextos, la interpretación es necesaria para confiar en el resultado. Si un modelo recomienda revisar ciertas transacciones como posibles fraudes, una persona puede necesitar entender qué señales motivaron esa alerta. Si un modelo estima riesgo en un contexto educativo, médico, financiero o institucional, las explicaciones se vuelven todavía más importantes.

Cuanto más sensible es la decisión, más importante es entender cómo se usa el modelo.

Interpretar no significa justificar cualquier resultado. Un modelo puede producir una predicción y aun así estar usando señales incorrectas, sesgadas o filtradas. Por eso, la interpretación también sirve para auditar. Nos ayuda a detectar si el modelo está aprendiendo algo razonable o si está aprovechando atajos problemáticos.

Por ejemplo, si un modelo para predecir aprobación de un examen usa una variable que solo se conoce después del examen, tendríamos fuga de datos. La métrica podría parecer excelente, pero la interpretación de variables revelaría que el modelo está usando información que no debería estar disponible al momento de predecir.

Una buena métrica no garantiza un buen modelo.

Interpretar ayuda a revisar qué está aprendiendo.

También puede ocurrir que una variable aparezca como muy importante, pero en realidad esté funcionando como sustituto de otra información sensible o problemática. Por ejemplo, una variable geográfica podría estar asociada a diferencias socioeconómicas. Una variable de comportamiento digital podría reflejar acceso desigual a conectividad. Interpretar exige mirar más allá del algoritmo.

Esto conecta con una idea que veremos con más detalle en el capítulo siguiente: los modelos no aprenden en el vacío. Aprenden desde datos que fueron producidos en un contexto social, técnico e institucional. Por eso, sus patrones deben analizarse con responsabilidad.

Interpretar modelos no es solo mirar números.

También es preguntarse si esos números tienen sentido en el contexto.

Hay distintos niveles de interpretación. A veces queremos entender el comportamiento global del modelo: qué variables parecen ser más importantes en general, qué relaciones aprendió, qué patrones predominan. Otras veces queremos interpretar una predicción individual: por qué este caso particular recibió cierta clase o cierto valor.

| Concepto | Descripción |
|---|---|
| Interpretación global | cómo se comporta el modelo en conjunto. |
| Interpretación local | por qué produjo una predicción particular. |

Por ejemplo, en un modelo de precios de viviendas, una interpretación global podría decir que la superficie, la ubicación y la cantidad de ambientes son variables importantes. Una interpretación local podría explicar por qué una vivienda específica recibió una predicción alta: tiene gran superficie, está en una zona de alto valor y posee varias habitaciones.

Ambos niveles son útiles, pero responden preguntas distintas. La interpretación global ayuda a comprender el modelo como sistema. La interpretación local ayuda a revisar casos concretos.

También debemos diferenciar interpretación de causalidad. Si un modelo aprende que cierta variable está asociada con una predicción, eso no significa automáticamente que esa variable cause el resultado. Un modelo predictivo puede usar correlaciones sin demostrar causas.

| Concepto | Descripción |
|---|---|
| Interpretación de modelo | qué señales usa para predecir. |
| Causalidad | qué factores producen realmente un efecto. |

Esta distinción es fundamental. Si un modelo encuentra que los clientes que llaman mucho al soporte abandonan más, eso no significa necesariamente que llamar al soporte cause abandono. Puede ser que los clientes llamen porque ya tienen problemas con el servicio, y esos problemas sean la verdadera causa del abandono.

El modelo puede usar la cantidad de reclamos como señal predictiva, pero la interpretación causal requiere otro tipo de análisis.

Una variable puede ser útil para predecir sin ser la causa del fenómeno.

Interpretar modelos también implica reconocer sus límites. Algunos modelos son más fáciles de interpretar que otros. Una regresión lineal o una regresión logística permiten observar coeficientes. Un árbol pequeño puede seguirse mediante reglas. En cambio, modelos como Random Forest, Gradient Boosting o SVM con kernels pueden ser más difíciles de explicar directamente.

Algunos modelos son más transparentes.

Otros son más potentes, pero menos fáciles de explicar.

A veces existe un equilibrio entre desempeño e interpretabilidad. Un modelo más complejo puede obtener mejores métricas, pero ser más difícil de comunicar. Un modelo más simple puede rendir un poco menos, pero ser más comprensible y suficiente para el objetivo.

No siempre necesitamos el modelo más interpretable. Tampoco siempre necesitamos el modelo más complejo. La decisión depende del problema, del riesgo de los errores, de las personas que usarán el resultado y de la necesidad de justificar decisiones.

Elegir un modelo también implica considerar qué tan explicable debe ser.

Por ejemplo, si usamos un modelo para organizar una lista interna de productos recomendados, quizá aceptemos mayor complejidad. Si lo usamos para tomar decisiones que afectan oportunidades de personas, la exigencia de interpretación y revisión debería ser mayor.

Interpretar también ayuda a comunicar resultados. Un informe de Machine Learning no debería limitarse a decir “el modelo obtuvo 0.87 de F1-score”. Conviene explicar qué variables parecen relevantes, qué errores se observaron, qué casos son difíciles y qué limitaciones tiene el análisis.

Comunicar un modelo: métricas, variables relevantes, errores, limitaciones, contexto.

Sin embargo, debemos evitar interpretar de más. No toda técnica de interpretación ofrece una verdad absoluta. La importancia de variables puede cambiar según el modelo, el dataset, el preprocesamiento y la métrica usada. Una explicación local puede ser aproximada. Un gráfico puede sugerir una relación, pero no demostrar una causa.

La interpretación debe ser útil, pero prudente.

Interpretar no es inventar certezas. Es construir explicaciones razonables con límites claros.

En este capítulo estudiaremos distintas formas de interpretar modelos. Veremos la diferencia entre modelos transparentes y modelos opacos, cómo leer coeficientes, cómo pensar la importancia de variables, qué aporta la permutation importance, y qué diferencia hay entre interpretaciones locales y globales. También dedicaremos atención a los riesgos de sobreinterpretar.

> Interpretar un modelo significa analizar cómo se comporta, qué variables usa, qué patrones parece haber aprendido y qué límites tienen esas explicaciones. La interpretación ayuda a confiar, auditar, comunicar y mejorar modelos. Pero debe hacerse con cuidado: una explicación de modelo no es necesariamente una explicación causal, ni una garantía de justicia, ni una verdad definitiva sobre el fenómeno. Es una herramienta para entender mejor el modelo y usarlo de manera más responsable.


Modelos transparentes y modelos opacos

No todos los modelos son igual de fáciles de interpretar. Algunos permiten observar con bastante claridad cómo relacionan las variables de entrada con la predicción. Otros pueden producir muy buenos resultados, pero su funcionamiento interno es más difícil de explicar. Por eso suele hablarse de modelos transparentes y modelos opacos.

Un modelo transparente es aquel cuya lógica puede entenderse de manera relativamente directa. Esto no significa que sea trivial, ni que no requiera conocimiento técnico, pero sí que sus decisiones pueden examinarse con cierta claridad.

Modelo transparente: su funcionamiento puede explicarse de forma relativamente directa.

La regresión lineal es un ejemplo clásico. En una regresión lineal, cada variable tiene un coeficiente. Ese coeficiente indica cómo cambia la predicción cuando aumenta esa variable, manteniendo las demás constantes. Si el modelo está bien planteado, podemos analizar esos coeficientes y construir una explicación razonable.

También ocurre algo parecido con la regresión logística. Aunque se use para clasificación, permite observar coeficientes asociados a las variables. Esos coeficientes ayudan a entender qué variables empujan la predicción hacia una clase u otra, siempre con cuidado en la interpretación.

| Concepto | Descripción |
|---|---|
| Regresión lineal | coeficientes asociados a cambios en un valor numérico. |
| Regresión logística | coeficientes asociados a la probabilidad de una clase. |

Un árbol de decisión pequeño también puede ser bastante transparente. Podemos seguir el camino desde la raíz hasta una hoja y observar qué preguntas hizo el modelo. Por ejemplo: si la antigüedad del cliente es menor que cierto valor, si tuvo reclamos recientes, si bajó su frecuencia de compra, y así sucesivamente.

Árbol pequeño: pregunta por una variable, divide los datos, vuelve a preguntar, llega a una decisión.

Esa estructura permite explicar una predicción como una secuencia de condiciones. Por eso, los árboles pequeños suelen ser útiles cuando necesitamos reglas comprensibles. Sin embargo, si el árbol crece demasiado, también puede volverse difícil de interpretar. La transparencia no depende solo del tipo de modelo, sino también de su tamaño y complejidad.

Un árbol pequeño puede ser claro.

Un árbol muy profundo puede ser difícil de seguir.

En el otro extremo aparecen modelos más opacos. Un modelo opaco, a veces llamado “caja negra”, es aquel cuyo comportamiento no puede explicarse fácilmente mirando directamente su estructura interna. Puede ser muy preciso, pero menos transparente.

Modelo opaco: puede predecir bien, pero su lógica interna no es fácil de explicar directamente.

Random Forest, por ejemplo, combina muchos árboles. Cada árbol puede tomar decisiones distintas y el resultado final surge de una votación o promedio. Aunque podamos analizar el modelo de distintas maneras, no es práctico explicar una predicción siguiendo manualmente cientos de árboles.

Gradient Boosting también puede ser difícil de interpretar directamente. Su predicción se construye como una secuencia de modelos que corrigen errores anteriores. El resultado puede ser muy potente, pero la explicación interna completa no suele ser sencilla.

| Concepto | Descripción |
|---|---|
| Random Forest | muchos árboles votando o promediando. |
| Gradient Boosting | muchos modelos pequeños corrigiendo errores paso a paso. |

SVM con kernels no lineales también puede resultar opaco. Aunque entendamos la idea del margen y los vectores de soporte, la frontera de decisión puede ser compleja. Explicar por qué un caso quedó de un lado u otro no siempre es directo.

Lo mismo ocurre con otros modelos más avanzados. A medida que aumenta la flexibilidad del modelo, muchas veces aumenta también la dificultad para explicar su comportamiento en términos simples.

Más flexibilidad puede traer mayor capacidad predictiva, pero menor interpretabilidad directa.

Esto no significa que los modelos opacos sean malos. Muchas veces son necesarios o convenientes. Si el objetivo principal es lograr el mejor desempeño posible en un problema complejo, un modelo opaco puede ser una buena elección. El problema aparece cuando usamos modelos difíciles de explicar en contextos donde la explicación es importante.

Por ejemplo, si un modelo se usa para priorizar campañas publicitarias, quizá la opacidad sea aceptable si el impacto de los errores es bajo y se monitorean los resultados. Pero si el modelo influye en decisiones educativas, financieras, médicas o legales, la necesidad de interpretación aumenta.

La exigencia de interpretación depende del riesgo y del uso del modelo.

También debemos evitar una idea demasiado simple: no siempre hay que elegir entre precisión e interpretabilidad como si fueran opuestos absolutos. A veces un modelo simple rinde casi tan bien como uno complejo. A veces un modelo complejo mejora mucho. A veces una pequeña pérdida de desempeño puede valer la pena si ganamos claridad, confianza y facilidad de comunicación.

La elección del modelo debe equilibrar: desempeño, interpretabilidad, costo, riesgo y contexto.

Supongamos que entrenamos dos modelos para predecir abandono de clientes. Una regresión logística obtiene un F1-score de 0.81. Un Gradient Boosting obtiene 0.83. La mejora existe, pero no es enorme. Si el equipo necesita entender y comunicar claramente los factores de abandono, quizá la regresión logística sea una opción suficiente y más conveniente.

En cambio, si el Gradient Boosting obtiene 0.90 y la mejora permite intervenir mucho mejor sobre clientes en riesgo, tal vez valga la pena aceptar mayor complejidad, siempre acompañada de técnicas de interpretación y monitoreo.

Un modelo más complejo debe justificar su complejidad con una mejora útil para el problema.

La transparencia también ayuda a detectar errores. Si un modelo interpretable muestra que una variable inesperada tiene mucho peso, podemos investigar. Tal vez haya fuga de datos. Tal vez la variable esté mal calculada. Tal vez represente un sesgo. En un modelo muy opaco, esos problemas pueden ser más difíciles de notar.

Pero tampoco debemos confiar ciegamente en los modelos transparentes. Que un modelo sea fácil de leer no significa que sea correcto, justo o causal. Una regresión logística puede usar variables problemáticas. Un árbol pequeño puede aprender una regla espuria. La interpretabilidad facilita la revisión, pero no reemplaza el criterio.

Transparente no significa automáticamente correcto o justo.

Los modelos opacos, por su parte, pueden analizarse mediante herramientas de interpretación. Podemos estudiar importancia de variables, permutation importance, gráficos parciales, explicaciones locales u otros métodos. Estas herramientas no vuelven al modelo completamente transparente, pero ayudan a comprender su comportamiento.

Un modelo opaco puede interpretarse parcialmente con herramientas adicionales.

Este punto es importante. La interpretabilidad no es una propiedad única y fija. Hay modelos naturalmente más transparentes, pero también existen técnicas para aproximarnos al comportamiento de modelos más complejos. Lo que cambia es el grado de esfuerzo, precisión y prudencia necesarios.

También debemos considerar a quién va dirigida la explicación. Una explicación técnica para un equipo de ciencia de datos puede incluir coeficientes, importancia de variables o curvas. Una explicación para una persona usuaria del sistema debe ser más clara, concreta y relacionada con el caso. Una explicación para una decisión institucional puede requerir documentación, validación y revisión ética.

No existe una única explicación.

La explicación adecuada depende de quién la necesita y para qué decisión.

Por ejemplo, decir “la variable X tiene alta importancia por permutation importance” puede ser útil para un equipo técnico. Pero para un informe general quizás convenga decir: “el modelo parece apoyarse principalmente en la actividad reciente, la antigüedad y la frecuencia de uso”. Para una persona afectada por una predicción, tal vez se necesite una explicación todavía más específica y cuidadosa.

La diferencia entre modelos transparentes y opacos, entonces, no debe verse como una clasificación rígida. Es más bien un continuo. Algunos modelos son muy fáciles de interpretar. Otros son parcialmente interpretables. Otros requieren herramientas externas. Otros pueden ser tan complejos que solo entendemos ciertos aspectos de su comportamiento.

Interpretabilidad: no es todo o nada.

Hay distintos grados de transparencia.

> Los modelos transparentes permiten examinar su lógica de manera más directa, como ocurre con regresiones o árboles pequeños. Los modelos opacos pueden ser más difíciles de explicar, como Random Forest, Gradient Boosting o SVM con kernels, pero a veces ofrecen mejor desempeño. La elección entre modelos debe considerar no solo métricas, sino también el riesgo del problema, la necesidad de explicación, el público destinatario y la posibilidad de auditar el comportamiento del modelo. Interpretar no es un lujo: en muchos proyectos, es parte de usar Machine Learning de manera responsable.


## Coeficientes

En algunos modelos, una de las formas más directas de interpretación aparece a través de los coeficientes. Los coeficientes son valores que indican cómo participa cada variable en la predicción del modelo. Son especialmente importantes en modelos lineales, como la regresión lineal y la regresión logística.

En una regresión lineal, el modelo construye una predicción sumando el aporte de distintas variables. Cada variable tiene asociado un coeficiente. Ese coeficiente indica cuánto cambia la predicción cuando esa variable aumenta una unidad, manteniendo las demás constantes.

El proceso parte de Variable, continúa con coeficiente y conduce finalmente a aporte a la predicción.

Por ejemplo, si entrenamos un modelo para predecir el precio de una vivienda, podríamos tener variables como superficie, cantidad de ambientes y antigüedad. Si el coeficiente de `superficie_m2` es positivo, el modelo está asociando una mayor superficie con un precio más alto.

| Concepto | Descripción |
|---|---|
| Coeficiente positivo | cuando la variable aumenta, la predicción tiende a aumentar. |
| Coeficiente negativo | cuando la variable aumenta, la predicción tiende a disminuir. |

Supongamos un ejemplo simplificado:

precio_estimado = 50000 + 1200 * superficie_m2.

En este caso, el coeficiente de `superficie_m2` es 1200. Eso significa que, según este modelo, por cada metro cuadrado adicional, el precio estimado aumenta en $1200, manteniendo todo lo demás constante.

Coeficiente de superficie_m2 = 1200

Interpretación: un metro cuadrado adicional se asocia con $1200 más en la predicción del modelo.

La expresión “manteniendo todo lo demás constante” es importante. En un modelo con varias variables, el coeficiente intenta describir el efecto de una variable cuando las demás no cambian. En la práctica, eso puede ser una simplificación fuerte, porque las variables muchas veces están relacionadas entre sí.

Por ejemplo, la superficie y la cantidad de ambientes suelen estar correlacionadas. Las viviendas más grandes tienden a tener más ambientes. Entonces, interpretar el coeficiente de una variable aislada puede ser más difícil cuando las variables están conectadas.

Coeficientes: útiles para interpretar, pero sensibles a la relación entre variables.

En regresión logística también existen coeficientes, pero su interpretación es menos directa que en regresión lineal. La regresión logística se usa para clasificación y estima probabilidades. Sus coeficientes indican cómo las variables empujan la predicción hacia una clase u otra, pero no representan un cambio directo en la probabilidad en unidades simples.

Aun así, el signo del coeficiente sigue siendo informativo. Un coeficiente positivo indica que, cuando esa variable aumenta, el modelo tiende a aumentar la evidencia a favor de la clase positiva. Un coeficiente negativo indica que la variable tiende a alejar la predicción de esa clase.

El proceso parte de Regresión logística: coeficiente positivo, continúa con más evidencia hacia la clase positiva, luego coeficiente negativo y conduce finalmente a menos evidencia hacia la clase positiva.

Por ejemplo, si el modelo predice abandono de clientes y `dias_desde_ultima_compra` tiene un coeficiente positivo, podría indicar que a mayor cantidad de días sin comprar, mayor tendencia del modelo a predecir abandono. Si `frecuencia_compra` tiene un coeficiente negativo, podría indicar que una mayor frecuencia de compra reduce la tendencia a predecir abandono.

El proceso parte de Más días sin comprar, continúa con mayor riesgo estimado de abandono, luego Mayor frecuencia de compra y conduce finalmente a menor riesgo estimado de abandono.

Esta interpretación puede ser muy útil para comunicar patrones generales. Pero debemos recordar que estamos interpretando el comportamiento del modelo, no una verdad causal. El modelo puede aprender una asociación entre días sin compra y abandono, pero eso no prueba por sí solo que los días sin compra causen el abandono.

Un coeficiente describe cómo usa una variable el modelo.

No demuestra por sí solo una causa.

En `scikit-learn`, los coeficientes de muchos modelos lineales se encuentran en el atributo `coef_`, después de entrenar el modelo.

```python id="y701ui"
modelo.coef_
```

En una regresión lineal con varias variables, `coef_` contiene un coeficiente por variable. Si tenemos los nombres de las columnas, podemos construir una tabla para interpretarlos mejor.

```python id="pbyh37"
import pandas as pd

coeficientes = pd.DataFrame({
    "variable": columnas,
    "coeficiente": modelo.coef_
})

coeficientes
```

En una regresión logística binaria, la estructura es similar, aunque `coef_` suele aparecer como una matriz. Podemos convertirla en una tabla de manera parecida.

```python id="q8yqfs"
coeficientes = pd.DataFrame({
    "variable": columnas,
    "coeficiente": modelo.coef_[0]
})

coeficientes.sort_values("coeficiente")
```

Ordenar los coeficientes puede ayudarnos a ver qué variables empujan más hacia una clase y cuáles empujan en sentido contrario. Pero esta comparación debe hacerse con mucho cuidado, especialmente si las variables están en escalas distintas.

Si una variable está medida en pesos y otra en unidades pequeñas, sus coeficientes no son directamente comparables. Un coeficiente pequeño en una variable con valores enormes puede tener un efecto importante. Un coeficiente grande en una variable binaria puede representar otra cosa. Por eso, la escala afecta la interpretación.

Comparar coeficientes solo tiene sentido si entendemos la escala de las variables.

Cuando las variables numéricas se estandarizan, los coeficientes pueden compararse de manera más razonable porque las variables quedan en escalas similares. Pero incluso en ese caso, debemos interpretar con prudencia.

Con variables escaladas, los coeficientes son más comparables.

Pero siguen siendo coeficientes del modelo, no explicaciones causales.

El preprocesamiento también afecta los coeficientes. Si usamos One-Hot Encoding para variables categóricas, una categoría puede transformarse en una columna binaria. El coeficiente de esa columna indica cómo cambia la predicción al pertenecer a esa categoría, en comparación con una categoría de referencia o con la codificación utilizada.

Por ejemplo, si una variable `medio_pago` se transforma en columnas como `medio_pago_tarjeta`, `medio_pago_transferencia` y `medio_pago_efectivo`, cada coeficiente se refiere a una de esas columnas transformadas, no a la variable categórica original como un todo.

Después de One-Hot Encoding, los coeficientes corresponden a columnas transformadas.

Esto puede hacer que la interpretación sea más detallada, pero también más extensa. En datasets con muchas categorías, la cantidad de coeficientes puede crecer mucho. Por eso, al interpretar modelos con variables categóricas codificadas, conviene revisar cuidadosamente cómo se construyeron las columnas.

También debemos prestar atención a variables correlacionadas. Cuando dos variables están muy relacionadas, el modelo puede repartir el peso entre ellas de una manera difícil de interpretar. A veces una variable parece tener poco coeficiente no porque no sea relevante, sino porque su información está compartida con otra.

Variables correlacionadas pueden dificultar la interpretación individual de coeficientes.

Por ejemplo, si usamos superficie, cantidad de ambientes y cantidad de dormitorios en un modelo de precio de viviendas, esas variables pueden estar muy relacionadas. El coeficiente de cada una puede depender de la presencia de las otras. Si quitamos una variable, los coeficientes de las restantes podrían cambiar.

Esto muestra que los coeficientes no deben leerse como atributos absolutos de las variables. Son parte de un modelo específico, entrenado con un conjunto específico de datos y variables.

Un coeficiente depende de: los datos, las variables incluidas, el preprocesamiento, y el modelo elegido.

La regularización también puede modificar los coeficientes. Como vimos en capítulos anteriores, Ridge y Lasso penalizan la magnitud de los coeficientes. Lasso incluso puede llevar algunos coeficientes a cero. Esto puede ayudar a controlar la complejidad, pero también afecta la interpretación.

Regularización: controla coeficientes y puede cambiar su magnitud.

Por eso, si interpretamos coeficientes de un modelo regularizado, debemos recordar que esos valores no solo reflejan asociaciones con la variable objetivo. También reflejan la penalización aplicada por el modelo.

Aun con estas precauciones, los coeficientes son una herramienta interpretativa muy valiosa. Permiten observar la dirección de la relación aprendida por el modelo y, en algunos casos, la magnitud relativa del aporte de cada variable.

Coeficientes útiles para: ver dirección, comparar aportes, detectar señales inesperadas, comunicar patrones generales.

Si una variable tiene un coeficiente con signo contrario al esperado, eso puede ser una pista para revisar los datos. Quizá haya un problema de codificación, una fuga de datos, una relación más compleja o una interpretación equivocada. Los coeficientes pueden funcionar como una herramienta de diagnóstico.

> Los coeficientes permiten interpretar modelos lineales observando cómo cada variable participa en la predicción. En regresión lineal, indican cambios en el valor predicho. En regresión logística, indican cómo las variables empujan la predicción hacia una clase u otra. Son útiles, pero deben leerse con cuidado: dependen de la escala, del preprocesamiento, de las variables incluidas, de la correlación entre ellas y de la regularización. Sobre todo, debemos recordar que un coeficiente describe el comportamiento del modelo, no una causa demostrada del fenómeno.


## Importancia de variables

Además de los coeficientes, otra forma frecuente de interpretar modelos consiste en analizar la importancia de variables. La idea general es identificar qué variables parecen tener mayor peso en el comportamiento del modelo. Dicho de manera simple: queremos saber qué columnas influyen más en las predicciones.

Importancia de variables: intento de estimar qué variables pesan más en el modelo.

Esta pregunta aparece en muchos proyectos. Si un modelo predice abandono de clientes, queremos saber si usa principalmente la actividad reciente, la antigüedad, la cantidad de reclamos, la frecuencia de compra o el monto gastado. Si predice precios de viviendas, queremos saber si influyen más la superficie, la ubicación, la cantidad de ambientes o la antigüedad. Si clasifica correos, queremos conocer qué términos ayudan más a distinguir spam de no spam.

La importancia de variables puede ayudar a comprender el modelo, comunicar resultados y detectar problemas. Si una variable aparece como muy importante, podemos investigarla. Si una variable que esperábamos relevante no aparece, también conviene revisar. Si una variable sospechosa domina el modelo, quizá exista fuga de datos o una señal poco apropiada.

Variables muy importantes: pueden revelar señales útiles, pero también posibles problemas.

Sin embargo, debemos tener cuidado desde el comienzo. “Importancia” no significa lo mismo en todos los modelos. Cada técnica puede calcularla de una manera distinta. Por eso, la importancia de variables no debe interpretarse como una verdad universal sobre los datos, sino como una descripción del comportamiento de un modelo específico.

Una variable puede ser importante para un modelo y menos importante para otro.

En modelos lineales, muchas veces usamos coeficientes para interpretar el aporte de cada variable. En árboles de decisión y ensambles basados en árboles, como Random Forest o Gradient Boosting, suelen existir medidas internas de importancia de variables. Estas medidas intentan estimar cuánto contribuye cada variable a mejorar las divisiones del modelo.

En `scikit-learn`, muchos modelos basados en árboles tienen el atributo `feature_importances_`.

```python id="mlcrdv"
modelo.feature_importances_
```

Si tenemos los nombres de las columnas, podemos construir una tabla para revisar esos valores.

```python id="d9p41y"
import pandas as pd

importancias = pd.DataFrame({
    "variable": columnas,
    "importancia": modelo.feature_importances_
})

importancias.sort_values("importancia", ascending=False)
```

El resultado puede mostrarnos, por ejemplo, que `dias_desde_ultima_compra`, `frecuencia_compra` y `cantidad_reclamos` son variables destacadas en un modelo de abandono. Eso nos da una pista sobre qué señales está usando el modelo.

Importancias altas: el modelo parece usar mucho esas variables para construir sus predicciones.

Pero estas importancias deben leerse con prudencia. En árboles, una variable puede aparecer como importante porque permite buenas divisiones en el entrenamiento. Eso no significa automáticamente que sea la causa del resultado. Tampoco significa que sea estable en otros datasets.

Por ejemplo, si una variable relacionada con reclamos aparece como importante para predecir abandono, podemos decir que el modelo la está usando como señal. No podemos afirmar solo con eso que los reclamos causan abandono. Puede ser que los reclamos sean consecuencia de problemas previos que también explican el abandono.

Importancia predictiva no equivale a importancia causal.

También puede haber variables correlacionadas. Si dos variables contienen información parecida, el modelo puede repartir la importancia entre ellas o preferir una por sobre la otra. En un Random Forest, por ejemplo, si `superficie_m2` y `cantidad_ambientes` están muy relacionadas, la importancia puede distribuirse de una manera que no representa claramente el valor conceptual de cada una.

Variables correlacionadas pueden complicar la interpretación de la importancia individual.

Si quitamos una variable, la otra podría ganar importancia. Si cambiamos el dataset o el modelo, la distribución puede cambiar. Por eso, no conviene leer la importancia de variables como un ranking definitivo e inmutable.

Otra precaución aparece con variables categóricas codificadas. Si usamos One-Hot Encoding, una variable categórica original puede convertirse en muchas columnas binarias. Por ejemplo, `barrio` puede transformarse en decenas de columnas, una por barrio. La importancia puede quedar repartida entre esas columnas.

| Concepto | Descripción |
|---|---|
| Variable categórica original | barrio |
| Después de One-Hot Encoding | barrio_A barrio_B barrio_C ... |

Si miramos cada columna por separado, quizá ninguna parezca demasiado importante. Pero, tomadas en conjunto, las columnas derivadas de `barrio` podrían tener mucho peso. Para interpretar correctamente, a veces necesitamos agrupar importancias por variable original.

También puede ocurrir lo contrario: una variable categórica con muchas categorías puede recibir mucha atención porque ofrece muchas oportunidades de división, aunque no necesariamente sea tan útil desde el punto de vista general. Algunas medidas internas de importancia en árboles pueden estar sesgadas hacia variables con más posibilidades de partición.

Las importancias internas pueden favorecer variables con muchas formas de dividir los datos.

Esto no significa que debamos ignorarlas, sino que debemos tomarlas como una primera aproximación. Son útiles para explorar, pero conviene complementarlas con otros análisis.

La importancia de variables también depende del preprocesamiento. Si escalamos, codificamos, imputamos o transformamos datos, el modelo ya no trabaja exactamente con las variables originales, sino con columnas transformadas. La interpretación debe tener en cuenta ese flujo.

El modelo interpreta las columnas que recibe, no necesariamente las variables originales tal como las pensamos.

Por ejemplo, si una variable numérica fue transformada, si se agregaron columnas indicadoras de valores faltantes o si una categoría fue expandida en muchas columnas, las importancias se refieren a esa representación final.

Por eso, cuando trabajamos con pipelines complejos, interpretar variables exige reconstruir qué columnas llegaron realmente al modelo. En proyectos simples esto puede ser directo. En proyectos con muchas transformaciones, puede requerir más cuidado.

Otra forma de usar la importancia de variables es como herramienta de revisión. Si una variable aparece con importancia muy alta y no debería estar disponible al momento de predicción, eso puede indicar fuga de datos.

El proceso parte de Variable muy importante, continúa con + no disponible al momento de predecir y conduce finalmente a posible fuga de datos..

Por ejemplo, en un modelo para predecir abandono, una variable como `fecha_baja` no debería usarse, porque solo se conoce después del abandono. Si aparece en el modelo, la evaluación estará contaminada. Una métrica alta en ese caso no sería confiable.

También puede ser sospechosa una variable que parezca una versión encubierta de la respuesta. Por ejemplo, `estado_final_cliente`, `motivo_cierre` o `resultado_revision` podrían contener información posterior al evento que queremos predecir.

La importancia de variables puede ayudarnos a detectar estos errores, pero no los detecta automáticamente. Necesitamos conocer el problema y revisar la disponibilidad temporal de cada variable.

Interpretar variables importantes también implica preguntar: ¿esta información existía en el momento de predicción?

Otro uso práctico es comunicar el modelo. En lugar de presentar solo métricas, podemos mostrar las variables que más influyen en las predicciones. Esto ayuda a que el análisis sea más comprensible para personas que no participaron del entrenamiento.

Pero la comunicación debe ser responsable. No conviene decir “estas son las causas del abandono” si lo que tenemos es una importancia predictiva. Es mejor decir “el modelo se apoya especialmente en estas variables” o “estas variables aparecen como señales relevantes para la predicción”.

| Concepto | Descripción |
|---|---|
| Más responsable | "variables relevantes para el modelo". |
| Menos responsable | "causas del resultado". |

También conviene acompañar la importancia con una explicación del contexto. Una variable puede ser importante porque contiene una señal legítima, porque está correlacionada con otra variable, porque refleja un sesgo o porque hay un problema de datos. La importancia por sí sola no distingue estas posibilidades.

> La importancia de variables busca identificar qué columnas parecen influir más en el comportamiento de un modelo. Puede ayudarnos a interpretar, comunicar, revisar y auditar. Pero debe usarse con prudencia: depende del modelo, del preprocesamiento, de las variables incluidas, de las correlaciones y de la forma en que se calcula. Una variable importante para predecir no es necesariamente una causa, ni un resultado universal. Es una pista sobre cómo está funcionando un modelo específico y debe analizarse con contexto.


## Permutation importance

La importancia de variables puede calcularse de distintas maneras. Algunas medidas vienen incorporadas en ciertos modelos, como ocurre con `feature_importances_` en árboles y ensambles. Pero también existe una técnica más general, que puede aplicarse a muchos tipos de modelos: la *permutation importance*, o importancia por permutación.

La idea es muy intuitiva. Si una variable es importante para el modelo, entonces desordenar sus valores debería empeorar el desempeño. Si una variable no aporta demasiado, desordenarla no debería cambiar mucho la métrica.

El proceso parte de Variable importante, continúa con si la desordenamos,, luego el modelo rinde peor., luego Variable poco importante, luego si la desordenamos, y conduce finalmente a el rendimiento cambia poco..

Imaginemos un modelo que predice abandono de clientes. Supongamos que una variable muy útil es `dias_desde_ultima_compra`. Si mezclamos al azar los valores de esa columna, cada cliente queda asociado con un valor que no le corresponde. El modelo pierde una señal importante y sus predicciones deberían empeorar.

En cambio, si desordenamos una variable poco relacionada con el abandono, el desempeño quizás cambie muy poco. Eso indicaría que el modelo no dependía demasiado de esa columna.

Permutation importance pregunta:

¿cuánto empeora el modelo cuando rompemos la información de una variable?

El procedimiento general es el siguiente. Primero evaluamos el modelo normalmente, usando una métrica elegida. Luego tomamos una variable, mezclamos sus valores entre las filas y volvemos a evaluar. La diferencia entre la métrica original y la métrica después de mezclar la variable indica su importancia.

1. Evaluar el modelo original.
2. Elegir una variable.
3. Mezclar sus valores.
4. Evaluar de nuevo.
5. Medir cuánto empeoró.

Si el desempeño cae mucho, la variable era importante para ese modelo en ese conjunto de datos. Si cae poco, la variable parecía tener poca influencia. Si no cambia, quizá el modelo no la estaba usando de manera relevante. Si incluso mejora, puede ser una señal de ruido, azar o inestabilidad del resultado.

Esta técnica tiene una ventaja importante: no depende de la estructura interna del modelo. Puede usarse con modelos lineales, Random Forest, Gradient Boosting, SVM, KNN u otros modelos, siempre que podamos medir su desempeño antes y después de permutar una columna.

Permutation importance observa el efecto sobre el rendimiento.

No necesita mirar cómo está construido internamente el modelo.

Por eso se considera una técnica relativamente general. Nos permite comparar la importancia de variables usando el mismo criterio: cuánto se deteriora la métrica cuando destruimos la información de una columna.

En `scikit-learn`, podemos usar `permutation_importance`.

```python id="bdby6e"
from sklearn.inspection import permutation_importance

resultado = permutation_importance(
    modelo,
    X_test,
    y_test,
    scoring="f1",
    n_repeats=10,
    random_state=42
)
```

En este ejemplo, la importancia se calcula sobre `X_test` y `y_test`, usando F1-score como métrica. El parámetro `n_repeats` indica cuántas veces se repite la permutación para cada variable, porque mezclar al azar puede producir pequeñas variaciones.

n_repeats: repite el proceso varias veces para obtener una estimación más estable.

El resultado incluye, entre otros valores, `importances_mean`, que contiene la importancia promedio de cada variable, e `importances_std`, que indica la variabilidad de esa estimación.

Podemos organizar esos valores en una tabla.

```python id="xtz695"
import pandas as pd

importancias = pd.DataFrame({
    "variable": columnas,
    "importancia_media": resultado.importances_mean,
    "desviacion": resultado.importances_std
})

importancias.sort_values(
    "importancia_media",
    ascending=False
)
```

Una variable con alta importancia media es aquella cuya permutación redujo mucho la métrica. La desviación ayuda a ver si esa importancia fue estable entre repeticiones. Si la desviación es alta, la estimación puede ser más incierta.

| Concepto | Descripción |
|---|---|
| Importancia media alta | al desordenar esa variable, el modelo pierde rendimiento. |
| Desviación alta | la estimación varía bastante entre repeticiones. |

La elección de la métrica es fundamental. La permutation importance no mide una importancia abstracta. Mide cuánto cambia una métrica concreta. Si usamos accuracy, obtendremos una lectura. Si usamos recall, podemos obtener otra. Si usamos F1-score, otra.

Esto es especialmente importante en problemas desbalanceados. Una variable puede no afectar demasiado la accuracy, pero sí afectar mucho el recall de la clase minoritaria. Por eso, la métrica elegida debe estar alineada con el objetivo del problema.

La importancia depende de la métrica usada.

Cambiar la métrica puede cambiar el ranking de variables.

Por ejemplo, en detección de fraude, quizá nos interese especialmente el recall de la clase fraude. Si calculamos permutation importance usando accuracy, podríamos subestimar variables relevantes para detectar casos raros. En cambio, una métrica enfocada en la clase de interés puede mostrar mejor qué variables sostienen esa capacidad del modelo.

Otra decisión importante es sobre qué datos calculamos la importancia. Conviene hacerlo sobre datos no usados para entrenar, como un conjunto de prueba o validación. Si calculamos la importancia sobre entrenamiento, podríamos medir cómo el modelo usa variables para ajustarse a datos ya vistos, no necesariamente cómo generaliza.

Mejor práctica: calcular importancia sobre datos no vistos o de validación.

Esto conecta con una idea que apareció muchas veces: no basta con que algo funcione en entrenamiento. Queremos entender qué variables sostienen el desempeño en datos nuevos.

Permutation importance también tiene limitaciones. Una de las más importantes aparece cuando las variables están correlacionadas. Si dos variables contienen información parecida, desordenar una sola puede no empeorar demasiado el modelo, porque la otra todavía conserva parte de la señal.

Variables correlacionadas: si desordenamos una, otra puede seguir aportando información similar.

Por ejemplo, en un modelo de precios de viviendas, `superficie_m2` y `cantidad_ambientes` pueden estar relacionadas. Si permutamos `superficie_m2`, el modelo quizá todavía use `cantidad_ambientes` como señal de tamaño. Entonces la importancia de `superficie_m2` podría parecer menor de lo que sería si estuviera sola.

Esto no significa que permutation importance esté mal. Significa que mide el efecto de romper una variable manteniendo las demás disponibles. Si hay variables redundantes, el resultado debe interpretarse con cuidado.

Permutation importance responde: ¿qué pasa si rompo esta columna y dejo las demás intactas?

Otra limitación es que puede ser costosa en tiempo. Para cada variable, el modelo debe evaluarse varias veces. Si tenemos muchas columnas, muchos datos o un modelo lento, el cálculo puede tardar. El parámetro `n_repeats` mejora la estabilidad, pero también aumenta el costo.

Además, cuando trabajamos con pipelines y preprocesamiento, debemos tener cuidado con los nombres de variables. Si el modelo recibe columnas transformadas, como variables escaladas o columnas generadas por One-Hot Encoding, la interpretación puede referirse a esas columnas transformadas, no siempre a las variables originales.

Con preprocesamiento complejo, la variable interpretada puede no coincidir directamente con una columna original.

En algunos casos, conviene calcular la importancia sobre el pipeline completo usando los datos originales de entrada. De esa manera, la permutación afecta las columnas originales antes del preprocesamiento. En otros casos, puede interesarnos interpretar columnas ya transformadas. La decisión depende de la pregunta.

Por ejemplo, si queremos saber si `barrio` como variable original es importante, quizás necesitemos agrupar el efecto de todas las columnas generadas por One-Hot Encoding. Si miramos cada categoría por separado, podemos perder la visión de conjunto.

El proceso parte de Variable original categórica, continúa con muchas columnas codificadas, luego importancia repartida y conduce finalmente a entre varias columnas..

También debemos recordar que permutation importance sigue siendo una técnica de interpretación del modelo, no una prueba causal. Si una variable tiene alta importancia, significa que el modelo depende de ella para sostener su rendimiento. No significa necesariamente que esa variable cause el resultado.

Alta permutation importance: la variable sostiene el rendimiento del modelo.

No significa automáticamente causalidad.

Por ejemplo, si `cantidad_reclamos` es importante para predecir abandono, el modelo usa esa señal. Pero eso no demuestra que los reclamos causen abandono. Puede haber factores previos que generen tanto reclamos como abandono.

A pesar de estas limitaciones, permutation importance es una herramienta muy útil. Permite interpretar modelos de distintos tipos con una lógica común. También ayuda a detectar variables sospechosas. Si al permutar una variable el rendimiento cae de manera enorme y esa variable no debería estar disponible al momento de predecir, puede haber fuga de datos.

El proceso parte de Variable demasiado importante, continúa con + disponibilidad sospechosa y conduce finalmente a revisar posible fuga de datos..

También puede ayudar a simplificar modelos. Si varias variables tienen importancia muy baja de manera estable, podríamos considerar si realmente aportan algo. Pero no conviene eliminarlas automáticamente. Antes de quitar variables, debemos evaluar el modelo completo y revisar si la eliminación afecta el desempeño.

Baja importancia no siempre implica eliminar la variable.

Primero hay que evaluar el efecto de quitarla.

> Permutation importance mide cuánto empeora el desempeño del modelo cuando desordenamos los valores de una variable. Si al romper una columna la métrica cae mucho, esa variable parece importante para el modelo. Es una técnica general, aplicable a muchos tipos de modelos, y especialmente útil para interpretar modelos más opacos. Sin embargo, depende de la métrica elegida, de los datos usados para evaluarla, de las correlaciones entre variables y del preprocesamiento. Como toda interpretación, debe leerse como una pista sobre el comportamiento del modelo, no como una explicación causal definitiva.


## Interpretación local y global

Cuando interpretamos un modelo, podemos hacerlo desde dos niveles distintos: interpretación global e interpretación local. Ambos niveles son útiles, pero responden preguntas diferentes.

La interpretación global intenta explicar el comportamiento general del modelo. Busca responder qué variables parecen importantes en conjunto, qué patrones predominan y cómo se comporta el modelo a nivel general.

Interpretación global: ¿cómo se comporta el modelo en términos generales?

Por ejemplo, en un modelo de abandono de clientes, una interpretación global podría mostrar que las variables más relevantes son los días desde la última compra, la frecuencia de compra, la cantidad de reclamos y la antigüedad. Esa lectura no explica un caso específico, sino el comportamiento general del modelo sobre muchos casos.

El proceso parte de Modelo de abandono, continúa con variables globalmente importantes: actividad reciente,, luego frecuencia,, luego reclamos, y conduce finalmente a antigüedad..

La importancia de variables y la permutation importance son ejemplos de herramientas de interpretación global. Nos ayudan a entender qué variables sostienen el desempeño del modelo en conjunto. También podemos usar gráficos, análisis de errores por grupo o comparaciones de métricas para obtener una mirada global.

La interpretación local, en cambio, intenta explicar una predicción individual. No pregunta qué hace el modelo en promedio, sino por qué produjo una predicción específica para un caso concreto.

Interpretación local: ¿por qué el modelo predijo esto para este caso particular?

Por ejemplo, un modelo puede predecir que un cliente tiene alto riesgo de abandono. La interpretación local intenta explicar qué características de ese cliente contribuyeron a esa predicción: muchos días desde la última compra, baja frecuencia reciente, reclamos abiertos o disminución en el uso del servicio.

| Concepto | Descripción |
|---|---|
| Predicción individual | cliente con alto riesgo. |
| Interpretación local | qué señales de ese cliente empujaron esa predicción. |

Esta diferencia es importante porque una variable puede ser muy importante a nivel global y no ser decisiva en todos los casos individuales. También puede ocurrir que una variable con importancia global moderada sea clave para una predicción específica.

Por ejemplo, en general, la frecuencia de compra puede ser una de las variables más importantes del modelo. Pero para un cliente particular, la predicción puede estar más influida por un reclamo reciente o por un cambio brusco en su patrón de uso.

Lo importante en promedio no siempre es lo que explica cada caso individual.

La interpretación global ayuda a comprender el modelo como sistema. La interpretación local ayuda a revisar decisiones puntuales. En muchos proyectos necesitamos ambas.

Supongamos que presentamos un modelo a un equipo de trabajo. A nivel global, podríamos explicar que el modelo se apoya principalmente en actividad reciente, frecuencia y reclamos. Pero si una persona pregunta por qué cierto cliente fue marcado como de alto riesgo, necesitamos una explicación local de ese caso.

| Concepto | Descripción |
|---|---|
| Global | qué patrones usa el modelo. |
| Local | qué ocurrió en una predicción concreta. |

En modelos transparentes, algunas interpretaciones locales pueden obtenerse de manera directa. En un árbol de decisión pequeño, podemos seguir el camino que recorrió un caso desde la raíz hasta la hoja final. Esa secuencia de preguntas explica por qué llegó a cierta predicción.

El proceso parte de Árbol de decisión: caso individual, continúa con camino de decisiones y conduce finalmente a predicción final.

En una regresión lineal o logística, podemos analizar cómo las variables del caso contribuyen al resultado según sus coeficientes. Si conocemos los valores de las variables y los coeficientes, podemos observar qué aportes empujan la predicción hacia arriba, hacia abajo o hacia una clase determinada.

En modelos más opacos, la interpretación local suele requerir herramientas adicionales. Existen técnicas diseñadas para aproximar qué variables influyeron en una predicción individual. No siempre son exactas en sentido absoluto, pero pueden ofrecer explicaciones útiles si se usan con cuidado.

Modelos opacos: la interpretación local requiere herramientas auxiliares y lectura prudente.

Una interpretación local puede ser especialmente valiosa cuando una predicción afecta una decisión concreta. Si un sistema marca una transacción como sospechosa, conviene saber qué señales motivaron la alerta. Si un modelo sugiere priorizar un caso para revisión, necesitamos saber qué información llevó a esa sugerencia.

Pero también puede ser riesgosa si se presenta como una explicación perfecta. Muchas técnicas locales son aproximaciones. Pueden depender del método, de los datos cercanos al caso, de la representación usada y de la estabilidad del modelo.

Una explicación local puede ayudar, pero no debe confundirse con una verdad definitiva.

También debemos recordar que una explicación local describe el comportamiento del modelo, no necesariamente la causa real del fenómeno. Si el modelo predijo alto riesgo de abandono porque hubo muchos días sin compra, eso indica que esa variable influyó en la predicción. No demuestra, por sí solo, la causa del abandono.

| Concepto | Descripción |
|---|---|
| Explicación local | por qué predijo el modelo. |
| No necesariamente | por qué ocurrió el fenómeno. |

La interpretación global también tiene límites. Un ranking de variables importantes puede ocultar diferencias entre subgrupos. El modelo puede apoyarse en ciertas variables para la mayoría de los casos, pero usar otras señales en grupos específicos. Si miramos solo el promedio, podemos perder esas diferencias.

Por ejemplo, un modelo de riesgo puede comportarse distinto con clientes nuevos y clientes antiguos. Una variable puede ser muy relevante para un grupo y poco relevante para otro. La interpretación global ofrece una visión general, pero no siempre captura todos los matices.

La interpretación global resume el comportamiento, pero puede ocultar diferencias entre subgrupos.

Por eso, a veces conviene combinar ambos niveles. Podemos empezar con una visión global para entender el modelo en conjunto y luego revisar casos individuales representativos: aciertos, errores, falsos positivos, falsos negativos, casos cercanos al umbral o casos de alto impacto.

Interpretar bien puede incluir: mirada global, casos individuales, errores, subgrupos, y contexto.

Los errores son especialmente importantes. Una interpretación global puede decirnos qué variables usa el modelo en general, pero revisar errores individuales puede mostrar en qué situaciones falla. Tal vez el modelo se confunde con clientes nuevos, con casos atípicos o con registros incompletos. Esos hallazgos pueden ser más útiles que mirar solo importancias promedio.

Revisar errores individuales ayuda a entender los límites reales del modelo.

También es útil analizar casos cercanos al umbral de decisión. En clasificación binaria, algunos casos tienen probabilidades cercanas al umbral elegido. Esos casos suelen ser más inciertos. Una explicación local puede ayudar a entender por qué quedaron de un lado u otro.

Casos cerca del umbral: decisiones más inciertas, más necesidad de revisión.

En cambio, casos muy alejados del umbral pueden ser más fáciles para el modelo. Pero incluso allí conviene revisar si la confianza aparente está justificada, especialmente si el modelo puede estar usando variables problemáticas.

A nivel práctico, la interpretación local y global deben estar alineadas con el uso del modelo. Si el modelo se usa para análisis general, quizá alcance con una interpretación global y ejemplos. Si se usa para decisiones individuales, la interpretación local se vuelve mucho más importante.

| Concepto | Descripción |
|---|---|
| Uso analítico general | más peso de la interpretación global. |
| Decisiones caso por caso | más peso de la interpretación local. |

También debemos adaptar la explicación al público. Un equipo técnico puede necesitar detalles sobre métricas, importancia de variables y métodos de interpretación. Una persona no técnica puede necesitar una explicación más concreta: qué factores principales influyeron y qué límites tiene la predicción.

> La interpretación global busca entender cómo se comporta el modelo en conjunto, mientras que la interpretación local intenta explicar predicciones individuales. La primera ayuda a comprender patrones generales, variables relevantes y comportamiento promedio. La segunda ayuda a revisar casos concretos, decisiones puntuales y errores específicos. Ambas son complementarias, y ambas deben interpretarse con prudencia: explican el comportamiento del modelo, no necesariamente las causas reales del fenómeno.


## Riesgos de interpretar de más

Interpretar modelos es necesario, pero también puede llevarnos a errores si convertimos las explicaciones en certezas demasiado fuertes. Un modelo puede ayudarnos a entender patrones, revisar variables importantes y comunicar resultados, pero sus interpretaciones tienen límites. Si olvidamos esos límites, corremos el riesgo de interpretar de más.

Interpretar de más significa sacar conclusiones que van más allá de lo que el modelo realmente permite afirmar.

| Concepto | Descripción |
|---|---|
| Interpretar bien | entender el comportamiento del modelo con prudencia. |
| Interpretar de más | convertir señales del modelo en conclusiones exageradas. |

Un primer riesgo es confundir importancia predictiva con causalidad. Si una variable aparece como importante, eso significa que el modelo la usa para predecir. No significa necesariamente que esa variable cause el resultado.

Por ejemplo, si un modelo de abandono de clientes muestra que la cantidad de reclamos es importante, podemos decir que el modelo usa esa variable como señal. Pero no podemos afirmar automáticamente que los reclamos causan el abandono. Puede ocurrir que los reclamos sean consecuencia de problemas previos en el servicio, y que esos problemas sean los que realmente explican la baja.

Variable importante para predecir no equivale a causa del fenómeno.

Esta diferencia es fundamental. Los modelos predictivos suelen aprender asociaciones. Algunas asociaciones son útiles para anticipar un resultado, pero no necesariamente explican por qué ocurre. Para hablar de causas necesitamos otros diseños de análisis, otros datos y otros criterios.

Un segundo riesgo es olvidar que la interpretación depende del modelo. Si entrenamos una regresión logística, un Random Forest y un Gradient Boosting sobre el mismo dataset, cada modelo puede apoyarse en las variables de manera diferente. Una variable puede ser muy importante en un modelo y menos importante en otro.

La interpretación pertenece a un modelo específico, entrenado con datos específicos.

Por eso, no conviene decir “esta variable es la más importante del problema” solo porque aparece primera en un ranking. Es más responsable decir “en este modelo, con estos datos y este preprocesamiento, esta variable aparece como muy relevante”.

La diferencia puede parecer pequeña, pero cambia mucho el sentido de la conclusión. La primera frase suena definitiva. La segunda deja claro que hablamos de un resultado situado.

| Concepto | Descripción |
|---|---|
| Más responsable | "En este modelo, esta variable parece relevante." |
| Menos responsable | "Esta es la variable más importante del fenómeno." |

Un tercer riesgo es no considerar el preprocesamiento. Los modelos no siempre trabajan con las variables originales tal como aparecen en la tabla. Pueden recibir columnas escaladas, imputadas, codificadas, combinadas o transformadas. Si interpretamos sin mirar ese flujo, podemos atribuir significado a columnas que ya no coinciden directamente con las variables iniciales.

Esto ocurre mucho con One-Hot Encoding. Una variable categórica puede transformarse en muchas columnas binarias. Si interpretamos cada columna por separado, podemos perder la visión de la variable original. También puede ocurrir que una categoría parezca importante, pero solo en relación con la forma en que se codificaron los datos.

El modelo interpreta la representación final de los datos.

Nosotros debemos reconstruir qué significa esa representación.

Un cuarto riesgo es interpretar rankings como si fueran exactos. Las importancias de variables suelen ordenarse de mayor a menor, pero ese orden no siempre debe leerse con precisión absoluta. Dos variables con valores muy parecidos pueden intercambiar posiciones si cambia la muestra, el entrenamiento, la métrica o el modelo.

Un ranking de variables puede orientar, pero no siempre indica diferencias exactas y estables.

Si una variable tiene importancia 0.21 y otra 0.20, no deberíamos construir una conclusión fuerte diciendo que la primera es claramente más importante. Puede ser una diferencia pequeña dentro de la variabilidad del análisis.

Por eso, cuando sea posible, conviene revisar estabilidad: repetir validaciones, comparar modelos, observar desviaciones en permutation importance o analizar si el patrón se mantiene en distintos subconjuntos de datos.

Interpretaciones más confiables suelen ser aquellas que se mantienen ante cambios razonables del análisis.

Un quinto riesgo es ignorar variables correlacionadas. Cuando dos columnas contienen información parecida, la importancia puede repartirse, ocultarse o desplazarse. Una variable puede parecer poco importante porque otra variable relacionada ya aporta la misma señal. O una puede aparecer como dominante solo porque el modelo la eligió antes.

Por ejemplo, en viviendas, superficie, ambientes y dormitorios pueden estar relacionadas. Si superficie aparece como la variable más importante, no significa que las demás no importen. Puede significar que superficie resume parte de la información que también contienen las otras.

Variables correlacionadas pueden hacer que la importancia sea difícil de repartir e interpretar.

Esto también afecta a los coeficientes. En modelos lineales, la presencia de variables correlacionadas puede cambiar la magnitud y el signo de los coeficientes. Si quitamos o agregamos una variable, los coeficientes pueden modificarse. Por eso, leerlos como efectos aislados puede ser engañoso.

Un sexto riesgo es interpretar patrones aprendidos de datos sesgados como si fueran patrones justos o neutrales. Si los datos reflejan desigualdades, exclusiones, errores históricos o decisiones previas, el modelo puede aprender esas estructuras. Interpretar el modelo sin revisar el origen de los datos puede llevarnos a normalizar esos sesgos.

Un modelo puede aprender patrones presentes en los datos, aunque esos patrones sean injustos o problemáticos.

Por ejemplo, si un modelo se entrena con datos de oportunidades pasadas y esas oportunidades estuvieron distribuidas de manera desigual, el modelo puede reproducir esa desigualdad. Una variable puede aparecer como predictiva, pero su uso puede ser cuestionable desde el punto de vista ético o institucional.

Por eso, interpretar no es solo mirar qué variable pesa más. También es preguntar si esa variable debería usarse, qué representa realmente y qué consecuencias puede tener.

| Concepto | Descripción |
|---|---|
| Pregunta técnica | ¿la variable ayuda a predecir? |
| Pregunta responsable | ¿es correcto usarla para esta decisión? |

Un séptimo riesgo es confiar demasiado en explicaciones locales. Una explicación local puede ayudarnos a entender una predicción individual, pero muchas técnicas locales son aproximadas. Pueden cambiar si modificamos el método de explicación, el conjunto de referencia o algunos detalles del modelo.

Esto no significa que no sirvan. Significa que deben leerse como ayudas interpretativas, no como una reconstrucción perfecta del razonamiento del modelo.

Una explicación local puede orientar una revisión, pero no siempre agota todo lo que hizo el modelo.

También puede ocurrir que una explicación local sea convincente, pero incompleta. Tal vez muestre las variables que más influyeron en una predicción, pero no indique que el modelo tiene bajo desempeño en cierto subgrupo. O puede explicar un caso particular sin revelar un problema global de sesgo o calibración.

Por eso, las explicaciones locales deben complementarse con análisis global, revisión de errores y evaluación por grupos cuando corresponda.

Buena interpretación: mirada global + casos individuales + errores + contexto.

Un octavo riesgo es interpretar probabilidades sin revisar calibración. Un modelo puede asignar una probabilidad alta a una clase, pero eso no significa necesariamente que esa probabilidad esté bien calibrada. Como vimos en otros capítulos, algunos modelos producen puntuaciones o probabilidades que requieren cuidado.

Si un modelo dice 0.90, no siempre significa que, de cada 100 casos similares, 90 pertenecerán realmente a esa clase. Puede ser una señal fuerte para el modelo, pero no una probabilidad confiable en sentido estricto.

Probabilidad alta no siempre significa certeza bien calibrada.

Esto es especialmente importante cuando las probabilidades se usan para tomar decisiones, fijar umbrales o comunicar riesgo. En esos casos, interpretar la salida del modelo requiere evaluar no solo la métrica de clasificación, sino también la calidad de las probabilidades.

Un noveno riesgo es olvidar que los modelos pueden cambiar con nuevos datos. Una interpretación válida hoy puede dejar de serlo si cambian los hábitos de los usuarios, las condiciones del mercado, los procesos de registro, las políticas de la organización o la forma en que se generan los datos.

La interpretación de un modelo puede envejecer si cambia el contexto que produce los datos.

Por eso, los modelos no deberían interpretarse una vez y considerarse cerrados para siempre. En proyectos reales, conviene monitorear desempeño, revisar errores y actualizar interpretaciones cuando el contexto cambia.

Un décimo riesgo es comunicar con un lenguaje demasiado fuerte. Palabras como “demuestra”, “prueba”, “causa” o “garantiza” pueden ser problemáticas si el análisis solo permite hablar de asociaciones o señales predictivas.

Es mejor usar expresiones más cuidadosas:

"El modelo sugiere..." "En estos datos se observa..." "Esta variable aparece asociada..." "El modelo se apoya en..." "Esta interpretación debe revisarse con...".

Ese tipo de lenguaje no debilita el análisis. Al contrario, lo vuelve más honesto y confiable. Un buen informe no exagera lo que el modelo sabe. Explica qué se encontró, cómo se obtuvo, qué límites tiene y qué revisiones serían necesarias.

> Interpretar modelos es valioso, pero interpretar de más puede llevar a conclusiones equivocadas. Los coeficientes, las importancias de variables, las explicaciones locales y los gráficos ayudan a entender el comportamiento del modelo, pero no demuestran causalidad ni garantizan justicia, estabilidad o verdad definitiva. Toda interpretación depende de los datos, el preprocesamiento, el modelo, la métrica y el contexto. Por eso, interpretar responsablemente significa explicar con claridad, reconocer límites y evitar afirmaciones más fuertes que la evidencia disponible.


## Interpretar para usar mejor

En este capítulo trabajamos una idea central: un modelo no debería evaluarse únicamente por sus métricas. Las métricas nos dicen cómo rinde bajo ciertas condiciones, pero no siempre nos dicen cómo está tomando sus decisiones, qué variables utiliza, qué patrones aprendió ni qué riesgos puede esconder. Para eso necesitamos interpretar.

Interpretar un modelo significa analizar su comportamiento. Queremos entender qué señales usa, qué variables parecen influir más, cómo se relacionan con las predicciones y qué límites tienen esas explicaciones.

Interpretar un modelo: entender cómo se comporta, qué señales usa y qué precauciones exige.

Vimos que no todos los modelos son igual de transparentes. Algunos, como regresiones lineales, regresiones logísticas o árboles pequeños, permiten una lectura relativamente directa. Otros, como Random Forest, Gradient Boosting o SVM con kernels, pueden ser más opacos y requieren herramientas adicionales para aproximarnos a su comportamiento.

| Concepto | Descripción |
|---|---|
| Modelos transparentes | más fáciles de explicar directamente. |
| Modelos opacos | pueden requerir herramientas de interpretación adicionales. |

También analizamos los coeficientes. En modelos lineales, los coeficientes permiten observar cómo participa cada variable en la predicción. En regresión lineal, pueden interpretarse como cambios en el valor predicho. En regresión logística, indican cómo las variables empujan la predicción hacia una clase u otra.

Pero los coeficientes deben leerse con cuidado. Su interpretación depende de la escala de las variables, del preprocesamiento, de la presencia de variables correlacionadas, de la regularización y del conjunto de variables incluidas. Además, un coeficiente describe el comportamiento del modelo, no una causa demostrada del fenómeno.

Coeficiente: señal de cómo usa una variable el modelo.

No equivale automáticamente a una explicación causal.

Otra herramienta importante fue la importancia de variables. Esta idea busca identificar qué columnas parecen pesar más en el modelo. Puede ayudarnos a comunicar, revisar y auditar. Si una variable aparece como muy importante, puede ser una señal útil, pero también puede revelar un problema, como fuga de datos o una variable que no debería estar disponible al momento de predecir.

Variable importante: puede ser una señal útil, o una pista para revisar posibles problemas.

Estudiamos también la permutation importance. Esta técnica mide cuánto empeora el desempeño del modelo cuando desordenamos los valores de una variable. Si la métrica cae mucho, esa variable parece importante para sostener el rendimiento. Es una herramienta flexible porque puede aplicarse a distintos tipos de modelos.

Permutation importance: romper una variable y medir cuánto cae el desempeño del modelo.

Sin embargo, también tiene límites. Depende de la métrica elegida, de los datos usados para calcularla, de la presencia de variables correlacionadas y del preprocesamiento. Como toda técnica de interpretación, debe leerse como una pista sobre el comportamiento del modelo, no como una verdad definitiva.

Después diferenciamos interpretación global e interpretación local. La interpretación global intenta explicar el comportamiento general del modelo: qué variables son relevantes en conjunto, qué patrones predominan, cómo parece funcionar el modelo en promedio. La interpretación local intenta explicar una predicción individual.

| Concepto | Descripción |
|---|---|
| Interpretación global | cómo se comporta el modelo en conjunto. |
| Interpretación local | por qué produjo una predicción para un caso concreto. |

Ambas miradas son complementarias. Una importancia global puede ayudarnos a entender el modelo como sistema, pero no explica necesariamente cada caso particular. Una explicación local puede ayudarnos a revisar una predicción concreta, pero no describe por sí sola todo el comportamiento del modelo.

También dedicamos atención a los riesgos de interpretar de más. Este punto es fundamental. Una variable importante no es necesariamente una causa. Una explicación local no es necesariamente una reconstrucción perfecta del razonamiento del modelo. Un ranking de variables no es una verdad absoluta. Una probabilidad alta no siempre está bien calibrada. Un modelo entrenado con datos sesgados puede aprender patrones problemáticos.

Interpretar responsablemente significa explicar, pero también reconocer límites.

Por eso, el lenguaje importa. Es preferible decir “el modelo se apoya en estas variables” antes que “estas variables causan el resultado”. Es más prudente decir “en esta representación se observa una separación” antes que “los grupos existen de manera definitiva”. Una buena interpretación no exagera lo que el modelo puede afirmar.

Interpretar modelos también ayuda a mejorar el trabajo técnico. Puede revelar variables mal construidas, señales inesperadas, fuga de datos, errores de preprocesamiento, sesgos, inestabilidad o diferencias entre subgrupos. No es solo una etapa de comunicación; también es una herramienta de revisión.

Interpretar sirve para: comprender, comunicar, auditar, detectar problemas y mejorar modelos.

> Interpretar un modelo es una parte esencial del trabajo con Machine Learning. Las métricas indican desempeño, pero la interpretación ayuda a entender qué está haciendo el modelo y si ese comportamiento tiene sentido. Coeficientes, importancia de variables, permutation importance e interpretaciones locales y globales son herramientas útiles, siempre que se usen con prudencia. Un modelo bien interpretado no es aquel al que le inventamos una explicación perfecta, sino aquel cuyo comportamiento podemos analizar, comunicar y revisar reconociendo sus límites.
